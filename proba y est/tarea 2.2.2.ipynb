# Tarea 2.2.2

### Supóngase que en cierta población el $1\%$ de las personas padece una enfermedad. Se dispone de una prueba diagnóstica con las siguientes características:

### a) si una persona está enferma, la prueba da positiva con probabilidad $0.95$;
### b) si una persona no está enferma, la prueba da positiva con probabilidad $0.08$.

### Si se elige una persona al azar y su prueba resulta positiva, ¿cuál es la probabilidad de que realmente esté enferma?

### Resolución:

Tenemos los siguientes eventos:

- Evento E: la persona padezca la enfermedad
- Evento S: la persona este sana
- Evento P: que la prueba resulte positiva

Notemos que $P(E) = 0.01$ es la probabilidad que la persona este enferma, además $P(S) = 1 - P(E)= 0.99$ es la probabilidad de que la persona este sana

Ahora el ejercicio nos pide calcular $P(E|P)$, es decir, la probabilidad que la persona este enferma dado que la prueba haya resultado positiva.

Sabemos que la $P(P|E) = 0.95$ es decir la probabilidad que la prueba de positiva dado que la persona padezca la enfermadad.

También notamos que $P(P|S) = 0.08$ es decir la probabilidad que la prueba de positiva dado que la persona no este enferma (osea que este sana)

Sabemos por teorema de Bayes que $P(E|P) = \frac{P(P|E) \cdot P(E)}{P(P)}$ ya tenemos $P(P|E) = 0.95$ y $P(E) = 0.01$, pero nos falta $P(P)$ que es la probabilidad que la prueba resulte positiva. 

Para calcular $P(P)$ usamos la probabilidad total, que sabemos que es:

$$P(B) = \sum_{i=1}^{n} P(B|A_i) \cdot P(A_i)$$

Pero para este caso, esta descrita como: 

$$P(P) = P(P|E) \cdot P(E) + P(P|S) \cdot P(S)$$

Es decir la probabilidad de que sea positiva es igual a la multiplicación de la probabilidad que la prueba sea positiva dado que la persona este enferma por la probabilidad que la persona este enferma, más la multiplicación de la probabilidad que la prueba sea positiva dado que la persona este sana por la probabilidad que la persona este sana, es decir, es la suma que de todas las posiblidades que puede suceder que una prueba de positivo.

Notamos que de esta forma ya podemos calcular $P(P)$, de la siguiente manera:

$$P(P) = P(P|E) \cdot P(E) + P(P|S) \cdot P(S) = (0.95)(0.01) + (0.08)(0.99)$$

Además recordemos que la probabilidad condicional está denotado como $P(A|B) = \frac{P(A \cap B)}{P(B)}$, despejando $P(A|B) \cdot P(B) = P(A \cap B)$, es decir, la probabilidad de A dado B multiplicado por B es la probabilidad de la intersección de A y B, adaptandolo a este ejercicio tendremos:

$P(P|E) \cdot P(E) = P(P \cap E)$: es la probabilidad de que una persona al azar este enferma y su prueba sea positiva (verdadero positivo)

$P(P|S) \cdot P(S) = P(P \cap S)$: es la probabilidad de que una persona al azar este sana y su prueba sea positiva (falso positivo)

```python
P_E = 0.01 # probabilidad que este enfermo
P_S = 1 - P_E # probabilidad que este sano

P_P_dado_E = 0.95 # probabilidad que sea positivo dado que está enfermo
P_P_dado_S = 0.08 # probabilidad que sea positivo dado que está sano

P_P = P_P_dado_E * P_E + P_P_dado_S * P_S # probabilidad que su prueba sea positivo

P_E_dado_P = (P_P_dado_E * P_E) / P_P # probabilidad que este enfermo dado que su prueba sea positiva

print(f'La probabilidad de que una persona este enferma dado que su prueba salio positivia es: {P_E_dado_P}, es decir, el {round(P_E_dado_P*100,2)}%')
```
