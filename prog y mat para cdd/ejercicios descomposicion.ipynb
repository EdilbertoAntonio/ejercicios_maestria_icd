### Comprobación de $u_1^tu_2 = v_1^tv_2 = 0$.

$$
u_1^t u_2 = 
\begin{bmatrix}
1 & 1
\end{bmatrix} 
\begin{bmatrix}
0 \\
1
\end{bmatrix} =
(0*1) + (1*1) = 1 \\
$$
Notamos que no es igual a $0$

$$
v_1^t v_2 = 
\begin{bmatrix}
1 & 2 & 2
\end{bmatrix} 
\begin{bmatrix}
0 \\
1 \\
1
\end{bmatrix} =
(0*1) + (2*1) + (2*1) = 4 \\
$$

De igual manera no es igual a $0$, entonces $u_1^tu_2 \neq v_1^tv_2 \neq 0$

```python
import numpy as np

S = np.array([[1, 2, 2],
            [1, 3, 3]])

rango = np.linalg.matrix_rank(S)
print(f"El rango de la matriz calculado es: {rango}")

print("\nDescomposición")

u1 = np.array([[1], [1]])
u2 = np.array([[0], [1]])

v1_t = np.array([[1, 2, 2]])
v2_t = np.array([[0, 1, 1]])


print("S = u1*v1_t + u2*v2_t")
S_reconstruida = (u1 @ v1_t) + (u2 @ v2_t)

print("Matriz S Original:")
print(S)
print("\nMatriz S Reconstruida:")
print(S_reconstruida)

print("\nOrtogonalidad")
producto_u = u1.T @ u2
producto_v = v1_t @ v2_t.T
print(f"Resultado de u1^t * u2: {producto_u.item()}")
print(f"Resultado de v1^t * v2: {producto_v.item()}")
```

$$

Encontramos valores propios:

$$
