### NUEVO EJERCICIO (similar al de la tarea 2.2.2)

Vamos a modificar las probabilidades para que veas cómo cambian los números en una enfermedad más común (por ejemplo, una gripe en invierno).

Prevalencia: El 15% de la población está enferma ($P(E) = 0.15$) y el 85% está sana ($P(S) = 0.85$).

Si estás enfermo: La prueba da positivo el 90% de las veces ($P(P\vert{}E) = 0.90$) y da negativo por error el 10% de las veces ($P(N\vert{}E) = 0.10$).

Si estás sano: La prueba da negativo el 95% de las veces ($P(N\vert{}S) = 0.95$) y da positivo por error el 5% de las veces ($P(P\vert{}S) = 0.05$).

```python
P_E = 0.15
P_S = 0.85

P_P_dado_E = 0.9
P_N_dado_E = 0.1

P_N_dado_S = 0.95
P_P_dado_S = 0.05
```

### Ejercicio 1:

Pregunta: Si eliges a una persona al azar en esta población y le haces la prueba, ¿cuál es la probabilidad de que el resultado sea negativo?

Respuesta: 

$$P(-) = P(-|E) \cdot P(E) + P(-|S) \cdot P(S)$$
Debido a que es la intersección de la todas las formas que se puede obtener la prueba negativa

```python
P_N = P_N_dado_E * P_E + P_N_dado_S * P_S
print(P_N)
```

### Ejercicio 2:

Pregunta: ¿Cuál es la probabilidad de elegir a una persona al azar y que resulte que está enferma y además su prueba salga negativa?


Respuesta: 

$$P(E \cap N) = P(E|N) \cdot P(N) = P(N|E) \cdot P(E) = P(N \cap E) $$

$P(E|N) = \frac{P(N|E) \cdot P(E)}{P(N)}$

```python
P_E_dado_N = (P_N_dado_E * P_E)/P_N
P_E_n_N = P_E_dado_N * P_N
P_N_n_E = P_N_dado_E * P_E

print(f'primera {P_E_n_N}, segunda {P_N_n_E}, son iguales? { P_E_n_N == P_N_n_E }')
```

### Ejercicio 3:

Pregunta: Si se elige a una persona al azar y su prueba resulta negativa, ¿cuál es la probabilidad de que realmente esté sana?


Respuesta:

$$P(S|N) = \frac{P(N|S) \cdot P(S)}{P(N)} $$

```python
P_S_dado_N = (P_N_dado_S * P_S)/P_N

print(P_S_dado_N)
```
