```python
import pandas as pd

link="http://cape.fcfm.buap.mx/cruz-suarez/Data/INEGI/SDEMT225/ENOE_SDEMT325.csv"
data=pd.read_csv(link, encoding="latin1")
```

```python
datos = data[data["emp_ppal"].isin([1, 2])].copy()
datos["sexo"] = datos["sex"].map({'1': "Hombre", '2': "Mujer"})
datos["empleo"] = datos["emp_ppal"].map({1: "Informal", 2: "Formal"})
datos[["sexo", "empleo"]].head()
```

```python
muestra = datos.sample(10, replace=True)
muestra.dropna()
```

```python
tabla_empleo=pd.crosstab(datos["empleo"], datos["sexo"], normalize=True)
tabla_empleo
```

```python
marginal_sexo = tabla_empleo.sum(axis=0)
print(marginal_sexo)
tabla_empleo.div(marginal_sexo, axis=1) # condicional respecto al sexo, probabilidad de empleo formal dado que es hombre
```

```python
marginal_empleo = tabla_empleo.sum(axis=1)
print(marginal_empleo)
tabla_empleo.div(marginal_empleo, axis=0)  # condicional respecto al trabajo,  probabilidad que sea hombre dado que tiene empleo formal
```
