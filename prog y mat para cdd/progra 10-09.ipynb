5.27 (Intro to Data Science: Duplicate Elimination and Counting Frequencies) Use a
list comprehension to create a list of 50 random values in the range 1 through 10. Use
NumPy’s unique function to obtain the unique values and their frequencies. Display the
results.

```python
import random
import numpy as np

values = [random.randint(1, 10) for _ in range(50)]

unique_values, frequencies = np.unique(values, return_counts=True)

print("Original list:")
print(values)

print("\nUnique values and their frequencies:")
for value, frequency in zip(unique_values, frequencies):
    print(f"{value}: {frequency}")
```

5.28 (Intro to Data Science: Survey Response Statistics) Twenty students were asked to
rate on a scale of 1 to 5 the quality of the food in the student cafeteria, with 1 being “awful”
and 5 being “excellent.” Place the 20 responses in a list

1, 2, 5, 4, 3, 5, 2, 1, 3, 3, 1, 4, 3, 3, 3, 2, 3, 3, 2, 5

Determine and display the frequency of each rating. Use the built-in functions, statis-
tics module functions and NumPy functions demonstrated in Section 5.17.2 to display
the following response statistics: minimum, maximum, range, mean, median, mode, vari-
ance and standard deviation.

```python
import statistics

responses = [
    1, 2, 5, 4, 3, 5, 2, 1, 3, 3,
    1, 4, 3, 3, 3, 2, 3, 3, 2, 5
]

unique_responses, frequencies = np.unique(responses, return_counts=True)

print("Frequency of each rating:")
for response, frequency in zip(unique_responses, frequencies):
    print(f"{response}: {frequency}")


print("\nStatistics:")
print(f"Minimum: {min(responses)}")
print(f"Maximum: {max(responses)}")
print(f"Range: {max(responses) - min(responses)}")
print(f"Mean: {statistics.mean(responses)}")
print(f"Median: {statistics.median(responses)}")
print(f"Mode: {statistics.mode(responses)}")
print(f"Variance: {statistics.pvariance(responses)}")
print(f"Standard Deviation: {statistics.pstdev(responses)}")


print("\nNumpy:")
print(f"Minimum: {np.min(responses)}")
print(f"Maximum: {np.max(responses)}")
print(f"Range: {np.max(responses) - np.min(responses)}")
print(f"Mean: {np.mean(responses)}")
print(f"Median: {np.median(responses)}")
print(f"Mode: {statistics.mode(responses)}")
print(f"Variance: {np.var(responses)}")
print(f"Standard Deviation: {np.std(responses)}")


```

5.29 (Intro to Data Science: Visualizing Survey Response Statistics) Using the list in
Exercise 5.28 and the techniques you learned in Section 5.17.2, display a bar chart show-
ing the response frequencies and their percentages of the total responses.

```python
import matplotlib.pyplot as plt

ratings = [1, 2, 3, 4, 5]
frequencies = [responses.count(rating) for rating in ratings]

percentages = [frequency / len(responses) * 100
               for frequency in frequencies]

for rating, frequency, percentage in zip(ratings, frequencies, percentages):
    print(f"Rating {rating}: {frequency} responses ({percentage:.1f}%)")

plt.bar(ratings, frequencies)

plt.xlabel("Food Quality Rating")
plt.ylabel("Frequency")
plt.title("Student Cafeteria Food Quality")

for rating, frequency, percentage in zip(ratings, frequencies, percentages):
    plt.text(rating, frequency, f"{percentage:.1f}%",
             ha="center", va="bottom")

plt.show()
```

RollDie.py

```python
import matplotlib.pyplot as plt
import numpy as np
import random
import seaborn as sns

rolls = [random.randrange(1, 7) for i in range(60000)]
values, frequencies = np.unique(rolls, return_counts=True)

title = f'Rolling a Six-Sided Die {len(rolls):,} Times'
sns.set_style('whitegrid')
axes = sns.barplot(x=values, y=frequencies, hue=values,
                   palette='bright', legend=False)

axes.set_title(title)
axes.set(xlabel='Die Value', ylabel='Frequency')


axes.set_ylim(top=max(frequencies) * 1.10)
for bar, frequency in zip(axes.patches, frequencies):
    text_x = bar.get_x() + bar.get_width() / 2.0
    text_y = bar.get_height()
    text = f'{frequency:,}\n{frequency / len(rolls):.3%}'
    axes.text(text_x, text_y, text,
        fontsize=11, ha='center', va='bottom')
```
