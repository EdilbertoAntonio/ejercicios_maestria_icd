```python
s = pd.Series([909976, 8615246, 2872086, 2273305], index=["Stockholm", "London", "Rome", "Paris"], name="Population")
s
```

```python
fig, axes = plt.subplots(1, 4, figsize=(12, 3.5))

s.plot(ax=axes[0], kind='line', title="line")
s.plot(ax=axes[1], kind='bar', title="bar")
s.plot(ax=axes[2], kind='box', title="box")
s.plot(ax=axes[3], kind='pie', title="pie")

fig.tight_layout()
fig.savefig("ch12-series-plot.pdf")
fig.savefig("ch12-series-plot.png")
```

```python
fig, axes = plt.subplots(1, 4, figsize=(12, 3.5))

s.plot(ax=axes[0], kind='line', title="line", ylabel='Population')
s.plot(ax=axes[1], kind='bar', title="bar", ylabel='Population')
s.plot(ax=axes[2], kind='box', title="box")
s.plot(ax=axes[3], kind='pie', title="pie", label='Cities')
plt.legend()
```
