```python
covid = pd.read_csv('datos//covid-19.csv')
covid.head()

covid_conf = covid[covid['CLASIFICACION_FINAL'].isin([1, 2, 3])].copy()

comorbilidades = ['DIABETES', 'EPOC', 'ASMA', 'INMUSUPR', 'HIPERTENSION', 'OTRA_COM', 'CARDIOVASCULAR', 
                    'OBESIDAD', 'RENAL_CRONICA', 'TABAQUISMO', 'NEUMONIA']
columnas = comorbilidades + ['INTUBADO', 'TIPO_PACIENTE']
covid_asociacion = covid_conf[columnas].copy()

covid_asociacion.loc[(covid_asociacion['TIPO_PACIENTE'] == 1) & (covid_asociacion['INTUBADO'] == 97), 'INTUBADO'] = 2


covid_asociacion = covid_asociacion[covid_asociacion[comorbilidades + ['INTUBADO']].isin([1, 2]).all(axis=1)]

#
covid_asociacion = covid_asociacion[covid_asociacion['TIPO_PACIENTE'] == 2]

covid_bool = (covid_asociacion[comorbilidades + ['INTUBADO']] == 1)
covid_bool
```

```python
# frequent_fp = fpgrowth(covid_bool, min_support=0.001, use_colnames=True)
frequent_fp = fpgrowth(covid_bool, min_support=0.01, use_colnames=True)
frequent_fp
```

```python
rules = association_rules(frequent_fp, metric='confidence', min_threshold=0.05)
rules
```

```python
from scipy.stats import chi2_contingency, pearsonr
import numpy as np
```

```python
# 1. Asegúrate de tener ambas variables limpias (1 = Sí/Hospitalizado, 0 = No/Ambulatorio)
# Recuerda que para ambulatarios (TIPO_PACIENTE == 1), INTUBADO == 97 equivale a 0 (No intubado)
df_dep = covid[['TIPO_PACIENTE', 'INTUBADO']].copy()
df_dep.loc[(df_dep['TIPO_PACIENTE'] == 1) & (df_dep['INTUBADO'] == 97), 'INTUBADO'] = 2

# Filtrar solo códigos válidos (1 y 2)
df_dep = df_dep[df_dep['TIPO_PACIENTE'].isin([1, 2]) & df_dep['INTUBADO'].isin([1, 2])]

# Binarizar: 1 -> 1 (Hospitalizado / Intubado), 2 -> 0 (Ambulatorio / No intubado)
tipo_bin = (df_dep['TIPO_PACIENTE'] == 2).astype(int)
intu_bin = (df_dep['INTUBADO'] == 1).astype(int)

# 2. Tabla de contingencia 2x2
tabla = pd.crosstab(tipo_bin, intu_bin, rownames=['Hospitalizado'], colnames=['Intubado'])
print("Tabla de contingencia:\n", tabla, "\n")

# 3. Prueba Chi-cuadrada de independencia
chi2, p_valor, gl, esperados = chi2_contingency(tabla)
n = tabla.sum().sum()
phi = np.sqrt(chi2 / n)  # En tablas 2x2, V de Cramér == Phi

print(f"Chi-cuadrada: {chi2:.4f} | p-valor: {p_valor:.4e} | gl: {gl}")
print(f"Coeficiente Phi (fuerza de asociación): {phi:.4f}")

# 4. Comparación con Pearson sobre variables binarias (da exactamente |Phi|)
r_pearson, _ = pearsonr(tipo_bin, intu_bin)
print(f"Pearson binario (r): {r_pearson:.4f}")
```
