## 5.7

```python
def uniques(lista):
    unicos = []
    for elemento in lista:
        if lista.count(elemento) == 1:
            unicos.append(elemento)
        else:
            if elemento not in unicos:
                unicos.append(elemento)

    return sorted(unicos)

prueba_num = [2,2,4,5,6,7,6,5,1,8,1,8,1,1,3]
print(uniques(prueba_num))

prueba_str = 'dfgdfajjjkeqqqwmssm'
print(uniques(prueba_str))
```

```python
import time

n = 35
trues = [True]*n

inicio = time.perf_counter()

for i in range(2,n+1):
    for j in range(i+1,n+1):
        if j%i == 0:
            trues[j-1] = False

fin = time.perf_counter()
tiempo_transcurrido = fin - inicio
print(f"Tiempo de ejecución: {tiempo_transcurrido:.6f} segundos")
```
