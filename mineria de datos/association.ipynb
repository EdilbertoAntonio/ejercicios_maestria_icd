```python
transations = [['bread', 'milk', 'butter'], ['bread', 'milk'], 
                ['bread', 'butter'], ['milk', 'eggs'],['bread', 'milk', 'butter']]

transations
```

```python
import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, fpgrowth, association_rules

encoder = TransactionEncoder()
encoded = encoder.fit_transform(transations)
data1 = pd.DataFrame(encoded, columns = encoder.columns_)
data1
```

```python
frequent_apriori = apriori(data1, min_support = 0.6, use_colnames = True)
frequent_apriori
```

```python
association_rules(frequent_apriori, metric='confidence', min_threshold=0.7)
```

```python
frequent_fp = fpgrowth(data1, min_support=0.6, use_colnames=True)
frequent_fp
```

```python
association_rules(frequent_fp, metric='confidence', min_threshold=0.7)
```

```python
# determine the time difference between fpgrowth vs apriori. 
# perf_counter()
from time import perf_counter

start = perf_counter()
frequent_apriori = apriori(data1, min_support = 0.6, use_colnames = True)
apriori_time = perf_counter() - start

start = perf_counter()
frequent_fp = fpgrowth(data1, min_support = 0.6, use_colnames = True)
fp_growth_time = perf_counter() - start

print('apriori time ', apriori_time*1000)
print('fp growth time ', fp_growth_time*1000)
```
