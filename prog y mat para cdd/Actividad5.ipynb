## 5.7

```python
def uniques(lista):
    unicos = []
    for elemento in lista:
        if lista.count(elemento) == 1:
            unicos.append(elemento)
        else:
            if elemento not in unicos:
                unicos.append(elemento)

    return sorted(unicos)

prueba_num = [2,2,4,5,6,7,6,5,1,8,1,8,1,1,3]
print(uniques(prueba_num))

prueba_str = 'dfgdfajjjkeqqqwmssm'
print(uniques(prueba_str))
```

## 5.8

```python
n = 1000
primo = [True]*(n+1)
primo[0] = primo[1] = False 
limite = int(n**0.5) + 1

for i in range(2, limite):
    if primo[i] == True: #nos ayuda a saber si el número que vamos a verificar ya esta marcado o no con False
        for j in range(i * 2, n + 1, i):
            primo[j] = False

is_prime = list(enumerate(primo))

for i in range(n):
    if is_prime[i][1] == True:
        print(f'El número {is_prime[i][0]} es primo')
```

```python
dos = ['A', 'B', 'C']
tres = ['D', 'E', 'F']
cuatro = ['G', 'H', 'I']
cinco = ['J', 'K', 'L']
seis = ['M', 'N', 'O']
siete = ['P', 'R', 'S']
ocho = ['T', 'U', 'V']
nueve = ['X', 'Y', 'Z']

relacion = {'2': dos, '3': tres, '4': cuatro, '5': cinco, '6': seis, '7':siete, '8':ocho, '9':nueve}

def seven_letter(numero):
    if len(numero) != 7:
        return "El número no es de 7 digitos"

    if ('1' in numero) or ('0' in numero):
        return 'El 1 y 0 no están permitidos'
    
    palabras = []
    for l1 in relacion[numero[0]]:
        for l2 in relacion[numero[1]]:
            for l3 in relacion[numero[2]]:
                for l4 in relacion[numero[3]]:
                    for l5 in relacion[numero[4]]:
                        for l6 in relacion[numero[5]]:
                            for l7 in relacion[numero[6]]:
                                palabra = l1+l2+l3+l4+l5+l6+l7
                                palabras.append(palabra)
    
    return palabras

numero = str(input('Escribe un número de 7 digitos, pero que no incluyan el 1 o 0'))
seven = seven_letter(numero)
print(seven[:10])
            

```

```python
def is_ordered(sequence):
    return list(sequence) == sorted(sequence)

print(is_ordered([1,2,4,7,5,6,9]))
print(is_ordered([1,2,5,8,9,11,46]))
print(is_ordered((8,3,6,7,2,4)))
print(is_ordered((1,2,3,4,5,6,7,8,9)))
print(is_ordered('asfjdiso'))
print(is_ordered('abcdefgh'))
```
