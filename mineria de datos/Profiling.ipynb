<h1>Data profiling</h1>
   <div style="text-align: justify"> Data profiling is the process of reviewing and cleaning data to understand how data’s structure and maintain data quality standards. The main purpose is to gain insight into the quality of the data by using methods to review and summarize it, and then evaluating its condition.
Data profiling evaluates data based on factors such as accuracy, consistency, and timeliness to show if the data is lacking consistency or accuracy or has null values. Data profiling can be an important precursor to data processing and data analytics.<br>
Pandas Profiling is used to generate a complete and exhaustive report for the dataset, with many features and customizations in the generated report. This report includes various pieces of information such as dataset statistics, distribution of values, missing values, memory usage, etc., which are very useful for exploring and analyzing data efficiently.

<h1>Profiling Large Datasets</h1>
<div style="text-align: justify"> For small datasets, ydata-profiling computations can be performed quickly. However, for larger datasets, it may become too unwieldy.<br>
ydata-profiling includes a minimal configuration file where the most expensive computations are turned off by default. This configuration excludes time-consuming sections such as correlations and interactions.
</div>


```python
#Minimal configurations
report = ProfileReport(my_dat,minimal=True)
report
#profile.to_file("output.html")
```

An alternative way to **handle really large datasets** is to use a portion of it to generate the profiling report. Several users report this is a good way to scale back the computation time while maintaining representativity.

```python
#Sampling
sample = my_dat.sample(1000)
profile = ProfileReport(sample, minimal=True)
profile
```

<h1>Note on Primary and secondary data</h1>
<div style="text-align: justify">Data that has been collected from first-hand-experience is known as **primary data**. Primary data has not been published yet and is more reliable, authentic and objective. Primary data has not been changed or altered.<br>
Data collected from a source that has already been published in any form is called as **secondary data**. The review of literature in any research is based on secondary data. It is collected by someone else for some other purpose.
</div>

<h1>Note on correlation</h1>
Correlation is a statistical measure that expresses the extent to which two variables are linearly related (meaning they change together). It’s a common tool for describing simple relationships without making a statement about cause and effect.
Finding Relationships is a great aspect of the Pandas module. In Pandas we have the corr() method. The corr() method calculates the relationship between each column in your data set. Note: The corr() method ignores "not numeric" columns.<br>
The Result of the corr() gives numbers that vary from -1 to 1. 1 means that there is a 1 to 1 relationship (a perfect correlation), and for this data set, each time a value went up in the first column, the other one went up as well. -0.9 would be just as good relationship as 0.9, but if you increase one value, the other will probably go down.
Note: Correlation coefficients whose magnitude are between 0.9 and 1.0 indicate variables which can be considered very highly correlated. Correlation coefficients whose magnitude are between 0.7 and 0.9 indicate variables which can be considered highly correlated.
