### 2.8 

(Table of Squares and Cubes) Write a script that calculates the squares and cubes
of the numbers from 0 to 5. Print the resulting values in table format, as shown below. Use
the tab escape sequence to achieve the three-column output.

<table border="0">
  <tr>
    <th style="padding-right: 20px;">Number</th>
    <th style="padding-right: 20px;">Square</th>
    <th>Cube</th>
  </tr>
  <tr><td>0</td><td>0</td><td>0</td></tr>
  <tr><td>1</td><td>1</td><td>1</td></tr>

### 2.10 

(Arithmetic, Smallest and Largest) Write a script that inputs three integers from
the user. Display the sum, average, product, smallest and largest of the numbers. Note that
each of these is a reduction in functional-style programming.

```python
number1 = int(input('Ingrese el primer entero: '))
number2 = int(input('Ingrese el segundo entero: '))
number3 = int(input('Ingrese el tercer entero: '))

suma = number1 + number2 + number3
average = suma/3
product = number1 * number2 * number3
small = min(number1, number2 , number3)
largest = max(number1, number2 , number3)

print(f'la suma es: {suma}')
print(f'el promedio es: {average}')
print(f'el producto es: {product}')
print(f'el pequeño es: {small}')
print(f'el largo es: {largest}')
```

### 2.12

(7% Investment Return) Some investment advisors say that it’s reasonable to expect
a 7% return over the long term in the stock market. Assuming that you begin with
$1000 and leave your money invested, calculate and display how much money you’ll have
after 10, 20 and 30 years. Use the following formula for determining these amounts:

$a = p(1 + r)n$

where:

- p is the original amount invested (i.e., the principal of $1000),
- r is the annual rate of return (7%),
- n is the number of years (10, 20 or 30) and
- a is the amount on deposit at the end of the nth year.

```python
p = 1000
r = 0.07
years = [10,20,30]
a = [p*(1+r)**n for n in years]

for i in range(len(years)):
    print(f'en {years[i]} años tus $1000 se convertirán en {a[i]}')
```

### 2.15 

(Sort in Ascending Order) Write a script that inputs three different floating-point
numbers from the user. Display the numbers in increasing order. Recall that an if statement’s
suite can contain more than one statement. Prove that your script works by running
it on all six possible orderings of the numbers. Does your script work with duplicate
numbers? [This is challenging. In later chapters you’ll do this more conveniently and with
many more numbers.]

```python
number1 = float(input('Ingrese el primer flotante: '))
number2 = float(input('Ingrese el segundo flotante: '))
number3 = float(input('Ingrese el tercer flotante: '))

orden = []

if number1 <= number2 and number1 <= number3:
    if number2 <= number3:
        orden.extend([number1, number2, number3])
    else:
        orden.extend([number1, number3, number2])

elif number2 <= number1 and number2 <= number3:
    if number1 <= number3:
        orden.extend([number2, number1, number3])
    else:
        orden.extend([number2, number3, number1])

else:
    if number1 <= number2:
        orden.extend([number3, number1, number2])
    else:
        orden.extend([number3, number2, number1])

print(orden)
```
