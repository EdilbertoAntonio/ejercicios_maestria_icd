# 3. Variables Aleatorias

## 3.1 Definiciones Básicas

### Una variable aleatoria $X:\Omega \rightarrow \mathbb{R}$ definida en un espacio de probabilidad $(\Omega, \mathcal{F}, \mathbb{P})$ satisface la siguiente propiedad: 
$\begin{align}
X^{-1}(-\infty, x]\in \mathcal{F}, x\in \mathbb{R}.
\end{align}$

### De este modo se induce de manera natural una medida de probabilidad en el espacio medible $(\mathbb{R}, \mathcal{B}(\mathbb{R}))$, del modo siguiente:
$\begin{align}
F(x):=\mathbb{P}(X^{-1}(-\infty, x]), x \in \mathbb{R}. 
\end{align}$

### En el caso de variables aleatorias de tipo discreto, es más común describir su distribución a partir de su función puntual o de masa, definida como:

\begin{align}
p(x)=\mathbb{P}(X=x), x\in \mathcal{R}an(X)
\end{align}

### En el caso continuo la correspondeinte función es la densidad, la cual se obtiene también a partir de la distrbución acumulado, de acuerdo con:

\begin{align}
f(x)=\frac{dF(x)}{dx}.
\end{align}

### De este modo el cálculo de probabilidades se realiza del modo siguiente
$$
\mathbb P(X\in A)=
\begin{cases}
\displaystyle \sum_{x\in A} p(x), & \text{si } X \text{ es discreta},\\[1.2em]
\displaystyle \int_A f(x)\,dx, & \text{si } X \text{ es continua}.
\end{cases}
$$

### 3.2.1 Ejemplo 

### Sea $X\sim \text{Poisson}(\lambda)$, con $\lambda>0$. Su función de masa de probabilidad está dada por

### $$p(x)=P(X=x)=e^{-\lambda}\frac{\lambda^x}{x!}, \qquad x=0,1,2,\dots$$

### Además, esta función de masa puede escribirse de forma recursiva. En efecto, partiendo de

### $$p(0)=e^{-\lambda},$$

### se tiene que para $x\geq 0$,

### $$p(x+1)=\frac{\lambda}{x+1}\,p(x).$$

### Esta relación se obtiene al observar que

### $$p(x+1)=e^{-\lambda}\frac{\lambda^{x+1}}{(x+1)!}=\frac{\lambda}{x+1}\,e^{-\lambda}\frac{\lambda^x}{x!}=\frac{\lambda}{x+1}\,p(x).$$

### Esta forma recursiva nos permite calcular las probabilidades de manera iterativa sin necesidad de evaluar factoriales en cada paso. 

```python
import numpy as np

def poisson(lam, n):
    p=np.zeros(n+1)
    p[0]=np.exp(-lam)
    for x in range(n):
        p[x+1]=(lam/(x+1))*p[x]
    return p
```

```python
import pandas as pd

lam=3
n=10
masa=poisson(lam, n)
acumulada=np.cumsum(masa)

pd.DataFrame({
    'x': range(n+1),
    'p(x)': masa,
    'F(x)': acumulada
})
```

```python
import matplotlib.pyplot as plt

plt.stem(range(11), masa)
plt.xticks(range(11))
plt.show()
```

```python
plt.step(range(11), acumulada, where='post')
plt.show()
```

## 3.2.2 Ejemplo

Sea $X\sim \text{Exp}(\lambda)$, con $\lambda>0$. La distribución exponencial es una de las distribuciones continuas más utilizadas para modelar tiempos de espera hasta que ocurre un evento.

Su función de densidad está dada por

$$
f(x)=\lambda e^{-\lambda x}, \qquad x\geq 0,
$$

y su función de distribución acumulada es

$$
F(x)=P(X\leq x)=1-e^{-\lambda x}, \qquad x\geq 0.
$$


```python
lam=2
x=np.linspace(0, 5, 100)

fx=lam * np.exp(-lam * x)
Fx=1 - np.exp(-lam * x)

pd.DataFrame({
    'x': x,
    'f(x)': fx,
    'F(x)': Fx
})
```

```python
plt.plot(x, fx)
plt.show()
```

```python
plt.plot(x,Fx)
```

```python
plt.step(x, Fx, where='post')
plt.show()
```

## Ejemplo 3.2.3

### En una planta embotelladora se analizan dos variables aleatorias.

1. ### Sea $X$ el número de botellas con tapa defectuosa en una muestra de 20 botellas. Supóngase que
$$
X\sim \text{Binomial}(20,0.05).
$$

2. ### Sea $Y$ el contenido de líquido de una botella, medido en mililitros. Supóngase que
$$
Y\sim N(500,8^2).
$$

### Se desea calcular:

### A) la probabilidad de que exactamente 2 botellas tengan tapa defectuosa;
### B) la probabilidad de que a lo más 1 botella tenga tapa defectuosa;
### C) la probabilidad de que el contenido esté entre 490 y 510 ml;
### D) Determinar un umbral de contenido tal que solo el 5% de las botellas lo exceda.


### SOLUCIÓN

```python
from scipy.stats import binom, norm

n=20
p=0.05

print('Solución A):')
print(binom.pmf(2, n, p))
print('Solución B):')
binom.cdf(1, n, p)
```

```python
mu=500
sigma=8

print('Solucion C):')
print(norm.cdf(510, mu, sigma) - norm.cdf(490, mu, sigma))
print('Solucion D):')
print(norm.ppf(0.95, mu, sigma))
```

## 3.3 Percentiles, Cuantiles e Intercuantiles

### Los percentiles son medidas de posición que permiten identificar puntos característicos de una distribución. El cuantil de orden $p$ de una distribución, con $0<p<1$, es el valor $x_p$ de la variable aleatoria $X$ tal que
$\begin{align}
p=P(X\le x_p).
\end{align}$

### Algunos cuantiles particulares de uso frecuente son los siguientes:

- ### El cuantil de orden $0.25$ corresponde al **primer cuartil** ($Q_1$), también llamado **percentil 25** ($P_{25}$).
- ### El cuantil de orden $0.5$ corresponde a la **mediana** o **segundo cuartil** ($Q_2$), también llamado **percentil 50** ($P_{50}$).
- ### El cuantil de orden $0.75$ corresponde al **tercer cuartil** ($Q_3$), también llamado **percentil 75** ($P_{75}$).

```python
from scipy.stats import expon

lam=0.5

probabilidades=[0.25, 0.50, 0.75]  
cuantiles=expon.ppf(probabilidades, scale=1/lam)

pd.DataFrame({
                'cuantil': [0.25, 0.5, 0.75], 
                'valor': cuantiles
            })
```

### El **rango intercuartílico** (IQR) es una medida de dispersión estadística que se define como la diferencia entre el tercer cuartil ($Q_3$) y el primer cuartil ($Q_1$). 

```python
from scipy.stats import geom

p=0.3
x=np.arange(1, 20)
fx=geom.pmf(x, p)

Q1=geom.ppf(0.25, p)
Q2=geom.ppf(0.50, p)  
Q3=geom.ppf(0.75, p)

plt.stem(x, fx)
plt.axvline(x=Q1, color='red', linestyle='--', label='Q1')
plt.axvline(x=Q2, color='green', linestyle='--', label='Q2')
plt.axvline(x=Q3, color='blue', linestyle='--', label='Q3')
plt.legend()
plt.xticks(x)
plt.show()
```

### La cola superior muestra que la distribución geométrica es asimétrica a la derecha. Aunque la mayor parte de la probabilidad se concentra en valores pequeños, todavía existe probabilidad de observar valores mayores, la cual decrece gradualmente conforme aumenta $x$.
