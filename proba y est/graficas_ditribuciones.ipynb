```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as st
```

```python
# plt.style.use('ggplot')
# plt.style.use("fivethirtyeight")
plt.style.use("seaborn-v0_8-notebook")
```

# Distribuciones discretas

## Bernoulli

```python
p_valores = np.linspace(0,1,5)
k = [0,1]

fig, axes = plt.subplots(1, 5, figsize=(10, 4), sharey=True)

for i, p in enumerate(p_valores):
    pmf_valores = st.bernoulli.pmf(k, p)
    
    # Usamos gráficos de barras para distribuciones discretas
    axes[i].bar(k, pmf_valores, color=['#ff9999', '#66b3ff'], edgecolor='black')
    axes[i].set_xticks(k)
    axes[i].set_xticklabels(['0 (Falla)', '1 (Éxito)'])
    axes[i].set_title(f'p = {p}')
    axes[i].set_ylim(0, 1.1)

axes[0].set_ylabel('Probabilidad')
plt.suptitle('Distribución de Bernoulli para distintas p', fontsize=14)
plt.tight_layout()
```

## Binomial

```python
p_valores = [round(x,2) for x in np.linspace(0.1,0.9,5)]
n_valores = [5, 10, 20, 30, 50]

fig, axes = plt.subplots(1, 5, figsize=(16, 4))
for i, n in enumerate(n_valores):
    x = np.arange(0, n + 1)
    
    for p in p_valores:
        pmf_valores = st.binom.pmf(x, n, p)
        axes[i].plot(x, pmf_valores, marker='o', markersize=4, linestyle='--', alpha=0.8, label=f'p={p}')
    
    axes[i].set_title(f'n = {n}')
    axes[i].set_xlabel('Número de éxitos (k)')
    
    if i == 0:
        axes[i].set_ylabel('Probabilidad')
        axes[i].legend()

plt.suptitle('Distribución Binomial para distintas combinaciones de n y p', fontsize=14)
plt.tight_layout()
```

## Geométrica

```python
p_valores = [0.1, 0.3, 0.5, 0.7, 0.9]
k = np.arange(1,15)

fig, axes = plt.subplots(1, 5, figsize=(15, 4), sharey=True)

for i, p in enumerate(p_valores):
    pmf_valores = st.geom.pmf(k, p)
    
    axes[i].stem(k, pmf_valores)
    axes[i].set_xticks([1, 5, 10, 15])
    axes[i].set_title(f'p = {p}')
    axes[i].set_ylim(0, 1)
    axes[i].set_xlabel('Ensayos (k)')
    axes[i].grid()

axes[0].set_ylabel('Probabilidad')
plt.suptitle('Distribución Geométrica para distintas p', fontsize=14)
plt.tight_layout()
```

## Poisson

```python
lambda_valores = [1, 3, 5, 10, 20]
k = np.arange(40)

colores = ['#e6194b', '#3cb44b', '#f58231', '#4363d8', '#911eb4']

fig, axes = plt.subplots(1, 5, figsize=(15, 4), sharey=True)

for i, lambdas in enumerate(lambda_valores):
    pmf_valores = st.poisson.pmf(k, lambdas)

    axes[i].plot(k, pmf_valores, marker='o', linestyle='--', color=colores[i], alpha=0.8)
    axes[i].set_xticks([0, 10, 20, 30])
    axes[i].set_title(f'λ = {lambdas}')
    axes[i].set_xlabel('Ocurrencias (k)')
    axes[i].grid()

axes[0].set_ylabel('Probabilidad')
plt.suptitle('Distribución de Poisson para distintos valores de λ', fontsize=14)
plt.tight_layout()
```

# Continua

## Uniforme 

```python
x = np.linspace(-1, 6, 1000)

# si b = 5 y a = 1 entonces loc = a, scale = b-a
pdf_valores = st.uniform.pdf(x, loc=1, scale=4)

fig, ax = plt.subplots(figsize=(8, 4))

ax.plot(x, pdf_valores, color='teal', lw=2)
ax.fill_between(x, pdf_valores, color='teal', alpha=0.3)

ax.set_title('Distribución Uniforme Continua U(1, 5)')
ax.set_xlabel('Valor (x)')
ax.set_ylabel('Densidad de Probabilidad (PDF)')
ax.set_xlim(-1, 6)
ax.set_ylim(0, 0.35)

plt.tight_layout()
```

## Exponencial

```python
lambda_valores = [0.5, 1, 2.5, 4, 5]
x = np.linspace(0, 6, 1000)

colores = ['#e6194b', '#3cb44b', '#f58231', '#4363d8', '#911eb4']

fig, axes = plt.subplots(1, 5, figsize=(15, 4), sharey=True)

for i, lambdas in enumerate(lambda_valores):
    pdf_valores = st.expon.pdf(x = x, scale = 1/lambdas)

    axes[i].plot(x, pdf_valores, color=colores[i], alpha=0.8)
    axes[i].set_xticks(np.arange(1,6))
    axes[i].set_title(f'λ = {lambdas}')
    axes[i].set_xlabel('Tiempo de espera')
    axes[i].grid()

axes[0].set_ylabel('Densidad de Probabilidad (PDF)')
plt.suptitle('Distribución exponencial para distintos valores de λ', fontsize=14)
plt.tight_layout()
```

## Normal

```python
dest_valores = [0.3, 0.5, 1, 2, 5]
x = np.linspace(-5, 5, 1000)

colores = ['#e6194b', '#3cb44b', '#f58231', '#4363d8', '#911eb4']

fig, axes = plt.subplots(1, 5, figsize=(15, 4), sharey=True)

for i, dest in enumerate(dest_valores):
    pdf_valores = st.norm.pdf(x = x, loc = 0, scale = dest)

    axes[i].plot(x, pdf_valores, color=colores[i], alpha=0.8)
    axes[i].set_xticks(np.arange(-4, 5, 2))
    axes[i].set_title(fr'$\mu$ = 0, $\sigma$  = {dest}')
    axes[i].set_xlabel('X')
    axes[i].grid()

axes[0].set_ylabel('Densidad de Probabilidad (PDF)')
plt.suptitle(r'Distribución normal para distintos valores de $\sigma$', fontsize=14)
plt.tight_layout()
```

## Gamma

```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as st

# Variamos el parámetro de forma (a) y mantenemos la escala fija en 1
a_valores = [1, 2, 3, 5, 9]
b_valores = [0.5, 1, 2]
x_gamma = np.linspace(0, 20, 1000) # Rango continuo
colores = ['#e6194b', '#3cb44b', '#f58231', '#4363d8', '#911eb4']

fig, axes = plt.subplots(1, 5, figsize=(15, 4), sharey=True)

for i, a in enumerate(a_valores):
    # En scipy, el parámetro 'a' representa el parámetro de forma (alpha)
    for j, b in enumerate(b_valores):
        pdf_gamma = st.gamma.pdf(x_gamma, a=a, scale=1/b)
    
        axes[i].plot(x_gamma, pdf_gamma, color=colores[j], lw=2, alpha=0.8, label=fr'$\beta$={b}')
        
    axes[i].set_title(fr'$\alpha = {a}$')
    axes[i].set_xlabel('X')
    axes[i].grid(alpha=0.4)
    axes[i].legend()
    # Quitamos los bordes superior y derecho para mayor limpieza
    axes[i].spines[['top', 'right']].set_visible(False)

axes[0].set_ylabel('Densidad (PDF)')
fig.suptitle(r'Distribución Gamma variando los parámetros $\alpha$ y $\beta$', fontsize=14)
fig.tight_layout()
```
