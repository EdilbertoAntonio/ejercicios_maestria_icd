### 3.11 

(Miles Per Gallon) Drivers are concerned with the mileage obtained by their automobiles.
One driver has kept track of several tankfuls of gasoline by recording miles driven
and gallons used for each tankful. Develop a sentinel-controlled-repetition script that
prompts the user to input the miles driven and gallons used for each tankful. The script
should calculate and display the miles per gallon obtained for each tankful. After processing
all input information, the script should calculate and display the combined miles per
gallon obtained for all tankfuls (that is, total miles driven divided by total gallons used).

```python
total_gallons = []
total_miles = []

while True:
    gallons = float(input('Enter the gallons used (-1 to end): '))
    if gallons == -1:
        break
    miles = float(input('Enter the miles driven:'))
    print(f'The miles/gallon for this tank was: {miles/gallons}')
    total_gallons.append(gallons)
    total_miles.append(miles)

print(f'The overall average miles/gallon was {sum(total_miles)/sum(total_gallons)}')
```

### 3.12 

(Palindromes) A palindrome is a number, word or text phrase that reads the same
backwards or forwards. For example, each of the following five-digit integers is a palindrome:
12321, 55555, 45554 and 11611. Write a script that reads in a five-digit integer
and determines whether it’s a palindrome. [Hint: Use the // and % operators to separate
the number into its digits.]

```python
# primera forma

number = str(input('Ingresa un numero de longitud 5'))
if len(number) == 5:
    reves = number[::-1]
    if reves == number:
        print('Es palindromo')
    else:
        print('No es palindromo')
else:
    print('No es longitud 5')
```

```python
# segunda forma

number = int(input('Ingresa un numero de longitud 5'))
if number > 0:
    digitos = []
    digitos_reves = []

    while number !=0:
        digito = number % 10
        number = number // 10
        digitos.insert(0, digito)
        digitos_reves.append(digito)

    if len(digitos) != 5:
        print('Tu numero no es de 5 digitos')
    elif digitos == digitos_reves:
        print('Es palindromo')
    else:
        print('No es palindromo')
else:
    print('Tu numero tiene que ser mayor que cero')
```

### 3.15 
(Challenge: Approximating the Mathematical Constant e) Write a script that estimates
the value of the mathematical constant e by using the formula below. Your script
can stop after summing 10 terms.

$e = 1 + \frac{1}{1!} + \frac{1}{2!} + \frac{1}{3!} + ...$

```python
def factorial(n):
    resultado = 1
    for i in range(1, n+1):
        resultado *= i
    return resultado


e = 0
n = 100

for i in range(0,n):
    e = e + 1/factorial(i)

print(e)
```

### 3.28 
(Intro to Data Science: Mean, Median and Mode) Calculate the mean, median and
mode of the values 9, 11, 22, 34, 17, 22, 34, 22 and 40. Suppose the values included another
34. What problem might occur?

```python
import statistics as st

numbers = [9, 11, 22, 34, 17, 22, 34, 22, 40]

def resultados(numbers):
    mean = sum(numbers)/len(numbers)
    # median2 = sorted(numbers)[int(len(numbers)/2)]
    median = st.median(numbers)
    mode = st.mode(numbers)
    return mean, median, mode

mean1, median1, moda1 = resultados(numbers)
print(f'La media es {mean1}')
print(f'La moda es {moda1}')
print(f'La mediana es {median1}\n')


numbers.append(34)
mean2, median2, moda2 = resultados(numbers)

print(f'La media es {mean2}')
print(f'La moda es {moda2}')
print(f'La mediana es {median2}')
```

Ahora al añadir otro 34 vemos que la moda ahora será bi-modal, es decir, tendremos 2 valores que se repetirán las mismas veces, es decir, 3 veces se repite el 34 y 3 veces se repite el 22. Además notamos que el promedio sube y la mediana se mantiene.
