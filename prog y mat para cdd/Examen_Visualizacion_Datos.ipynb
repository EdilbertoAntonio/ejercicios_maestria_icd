# Examen práctico: Análisis y visualización de datos con Python

**Asignatura:** Programación y Matemáticas en Ciencia de Datos  
**Herramientas:** Python, Pandas, Matplotlib y Seaborn

## Objetivo

En este examen se trabajará con conjuntos de datos reales. El objetivo principal es utilizar **Pandas para explorar y organizar los datos** y emplear **Matplotlib y/o Seaborn para construir visualizaciones que permitan responder preguntas concretas**.

También se utilizarán conceptos básicos de estadística descriptiva, principalmente **media, varianza y desviación estándar**.

### Requisitos generales de las gráficas

Todas las gráficas deberán incluir, cuando corresponda:

- Título descriptivo.
- Nombre de los ejes.
- Unidades de las variables, cuando estén disponibles.
- Leyenda cuando se representen varias variables o categorías.
- Tamaño y presentación que permitan leer claramente la información.
- Una breve interpretación de lo observado.

**Importante:** cuando no se indique explícitamente qué tipo de gráfica utilizar, el estudiante deberá seleccionar una visualización apropiada para responder la pregunta.
