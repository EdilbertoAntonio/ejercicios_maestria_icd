<h1>Exploratory Analysis (EDA)</h1>
<div style="text-align: justify">
Exploratory Data Analysis is a methodology employed by data scientists for analyzing and summarizing data sets. Exploratory data analysis involves delving into datasets, utilizing techniques such as data visualization, to uncover key characteristics. This approach aids in determining optimal ways to manipulate data sources, facilitating the extraction of valuable insights, pattern identification, anomaly detection, hypothesis testing, and assumption verification.<br>
Primarily utilized to unearth information beyond formal modeling or hypothesis testing, EDA enhances the comprehension of dataset variables and their interrelationships. Furthermore, it serves as a valuable tool for assessing the suitability of statistical techniques for data analysis.<br>
The primary aim of exploratory data analysis (EDA) is to examine data without forming premature assumptions. It serves to uncover evident errors, comprehend patterns, identify outliers or unusual occurrences, and unveil meaningful relationships among variables. By employing exploratory analysis, data scientists ensure the validity and relevance of their results to achieve specific business objectives. EDA also assists stakeholders in confirming that their inquiries are well-directed, addressing queries related to standard deviations, categorical variables, and confidence intervals. Following the completion of EDA and the derivation of insights, its findings can be leveraged for advanced data analysis or modeling, encompassing machine learning applications.<br>
There are four main categories of exploratory data analysis (EDA):
<h3>Univariate Non-Graphical:</h3>
This form of analysis deals with a single variable, focusing on describing the data and identifying existing patterns. It does not explore causes or relationships. It involves using statistical techniques like measures of central tendency (like the mean or median), measures of spread (like the range or standard deviation), and measures of shape (like skewness or kurtosis)<br>

<h3>Univariate Graphical:</h3>
Graphical methods are essential for obtaining a comprehensive understanding of the data. Common types include:<br>
- Stem-and-leaf plots: Illustrating all data values and the distribution shape.<br>
- Histograms: Bar plots representing the frequency or proportion of cases for various value ranges.<br>
- Box plots: Graphically presenting the five-number summary (minimum, first quartile, median, third quartile, and maximum).
<h3>Multivariate Non-Graphical:</h3>
Involves the analysis of data arising from more than one variable. Multivariate non-graphical EDA techniques typically reveal relationships between two or more variables through cross-tabulation or statistical methods. It can also include techniques like regression analysis or principal component analysis.
<h3>Multivariate Graphical:</h3>
Utilizes graphics to display relationships between two or more sets of data. A commonly used graphic is the grouped bar plot or bar chart, where each group represents one level of a variable, and each bar within a group signifies the levels of another variable.
</div>

<h2>Categhorical vs Categhorical</h2>
<div style="text-align:justify">Categorical data can be divided into groups:<br>
- Nominal: Female, Male, Employed, Unemployed<br>
- Ordinal:Educational Level, Level of Satisfaction, Degree of Comprehension<br>
The analysis of categorical data includes data organization using contingency tables, cross-tabulations, or two way tables and also providing a bi-variate percentage (or frequency) distribution.</div>

```python
#Chi-Squared test for independence
import pandas as pd
import numpy as np
import scipy.stats as stats
conventional = np.array([159,189,191,198,235,245,250,256,261,265,266,280,343,356,383,403,414,428,432])
Germ_Free = np.array([158,192,193,194,195,202,212,215,229,230,237,240,244,247,259,301,301,321,337,415,434,444,485,496,529,537,624,707,800])
print(len(conventional))
print(len(Germ_Free))
mice = pd.DataFrame({"days": np.append(Germ_Free,conventional), "conditions":np.append(np.repeat("germ_free",len(Germ_Free)), np.repeat("conventional",len(conventional))),"Sex":np.array(["male"]*24+["female"]*24)})
mice
crosstb=pd.crosstab(mice.conditions,mice.Sex)
#15 male  14 female  germ free
#10 male 9 female    conventional

#H0: Attribute  A and antribute B are independent
#H1: Attribute  A and antribute B are not independent

stats.chi2_contingency(crosstb)
# Creating barplot
#barplot = crosstb.plot.bar(rot=0)
```

<h2>Multivariate Analysis</h2>
It is an extension of bivariate analysis which means it involves multiple variables at the same time to find correlation between them. Multivariate Analysis is a set of statistical model that examine patterns in multidimensional data by considering at once, several data variable.

```python
#Scatter Matrix/Pair wise plots
import pandas as pd 
import seaborn as sns
my_dat=pd.read_csv("SSNMX_catalogo_20240101_20240207.csv",skiprows=4,skipfooter=7,engine="python")
```

```python
sns.set_theme(style="ticks")
sns.pairplot(my_dat,hue="Clasificacion")
```

```python
#minor:    m<4
#Light:   4<=m<4.9
#Moderate  5.0<m<5.9
#Strong:      6<m<6.9
#Major:       7<m<7.9
#Great:        8<m
my_dat["Clasificacion"]="hola"
my_dat["Clasificacion"][my_dat["Magnitud"]<4]="minor"
my_dat["Clasificacion"][(my_dat["Magnitud"]>=4)&(my_dat["Magnitud"]<=4.9)]="light"
my_dat["Clasificacion"][(my_dat["Magnitud"]>=5)&(my_dat["Magnitud"]<=5.9)]="moderate"
my_dat["Clasificacion"][(my_dat["Magnitud"]>=6)&(my_dat["Magnitud"]<=6.9)]="strong"
my_dat["Clasificacion"][(my_dat["Magnitud"]>=7)&(my_dat["Magnitud"]<=7.9)]="major"
my_dat["Clasificacion"][my_dat["Magnitud"]>8]="great"

```

<h2>HeatMap</h2>
Here we are using a heat map to check the correlation between all the columns in the dataset. It is a data visualisation technique that shows the magnitude of the phenomenon as colour in two dimensions. The values of correlation can vary from -1 to 1 where -1 means strong negative and +1 means strong positive correlation.<br>
<h3>Note on correlation</h3>
Correlation is a statistical measure that expresses the extent to which two variables are linearly related (meaning they change together). It’s a common tool for describing simple relationships without making a statement about cause and effect.
Finding Relationships is a great aspect of the Pandas module. In Pandas we have the corr() method. The corr() method calculates the relationship between each column in your data set. Note: The corr() method ignores "not numeric" columns.<br>
The Result of the corr() gives numbers that vary from -1 to 1. 1 means that there is a 1 to 1 relationship (a perfect correlation), and for this data set, each time a value went up in the first column, the other one went up as well. -0.9 would be just as good relationship as 0.9, but if you increase one value, the other will probably go down.
Note: Correlation coefficients whose magnitude are between 0.9 and 1.0 indicate variables which can be considered very highly correlated. Correlation coefficients whose magnitude are between 0.7 and 0.9 indicate variables which can be considered highly correlated.

```python
sns.heatmap(my_dat.corr())
```

```python
data = load_diabetes(as_frame=True, scaled=False)
diabetes = data.frame
display(diabetes.head())
# non graphical EDA for univariate
diabetes.describe(include='all')
```

```python
def summary_univariate(x:pd.Series) -> pd.Series:
    return pd.Series({'count':x.count(), 
                      'missing':x.isnull().sum(), 
                      'pct_missing':x.isnull().mean()*100, 
                      'mean':x.mean(),
                      'median':np.median(x),
                      'mode':stats.mode(x, keepdims=True).mode[0], #x.mode()
                      'variance':np.var(x),
                      'std':np.std(x),
                      'min':np.min(x),
                      'max':np.max(x),
                      'Q1':np.quantile(x,0.25),
                      'Q3':np.quantile(x,0.75),
                      'IQR':np.quantile(x,0.75)-np.quantile(x,0.25),
                      'Skewness':x.skew(),
                      'Kurtosis':x.kurtosis()})

# example
summary_univariate(diabetes['age'])
```
