# 7. Estadística Inferencial

## 7.1 Introducción y Conceptos Básicos

### La **estadística inferencial** es la rama de la estadística que permite **extraer conclusiones sobre una población** a partir de la información obtenida de una **muestra**. En muchas situaciones no es posible observar a todos los elementos de interés, ya sea por costo, tiempo o dificultad de acceso. Por ello, se estudia una parte representativa de la población y, con base en ella, se realizan generalizaciones.

### Mientras que la **estadística descriptiva** se enfoca en organizar, resumir y presentar datos mediante tablas, gráficas y medidas numéricas, la **estadística inferencial** va un paso más allá: utiliza esos datos para **estimar parámetros**, **evaluar hipótesis** y **tomar decisiones** bajo incertidumbre.

### Por ejemplo, si se desea conocer el ingreso promedio de todos los hogares de una ciudad, no siempre es factible encuestarlos a todos. En su lugar, se selecciona una muestra de hogares y, a partir de ella, se estima el promedio poblacional.

### **Población**
### Es el conjunto total de elementos o individuos sobre los cuales se desea obtener información.

### **Muestra**
### Es un subconjunto de la población seleccionado para el estudio.

### **Parámetro**
### Es una característica numérica de la población, como la media $\mu$, la varianza $\sigma^2$ o la proporción $p$.

### **Estadístico**
### Es una medida calculada a partir de la muestra, como la media muestral $\bar{X}$, la varianza muestral $S^2$ o la proporción muestral $\hat{p}$.

### Al trabajar con muestras, las conclusiones pueden variar de una muestra a otra. Esta variabilidad se conoce como **error muestral**. Por ello, la estadística inferencial no ofrece certezas absolutas, sino conclusiones con cierto **nivel de confianza** o **margen de error**.

## 7.2 Definiciones

### Formalizando las definiciones anteriores, se tiene una v.a. $X \sim f_{\theta}$ que representa la variable de interés a observar en una población. Se realiza un muestreo de la población y se obtienen las siguientes observaciones: $X_1, X_2, \ldots, X_n$. Con base en ellas se propone un estadístico

### $$\hat{\theta} = h(X_1, X_2, \ldots, X_n).$$

### A continuación se muestra una tabla con algunos estadísticos muestrales comunes y los parámetros poblacionales que suelen estimar.
| Estadístico muestral | Expresión | Parámetro poblacional que estima |
|---|---|---|
| Media muestral | $\bar{X} = \dfrac{1}{n}\sum_{i=1}^{n} X_i$ | Media poblacional $\mu$ |
| Proporción muestral | $\hat{p} = \dfrac{X}{n}$ | Proporción poblacional $p$ |
| Varianza muestral | $S^2 = \dfrac{1}{n-1}\sum_{i=1}^{n}(X_i-\bar{X})^2$ | Varianza poblacional $\sigma^2$ |
| Desviación estándar muestral | $S = \sqrt{S^2}$ | Desviación estándar poblacional $\sigma$ |
| Mediana muestral | $\tilde{X} = \begin{cases} X_{\left(\frac{n+1}{2}\right)}, & \text{si } n \text{ es impar}, \\[6pt] \dfrac{X_{\left(\frac{n}{2}\right)} + X_{\left(\frac{n}{2}+1\right)}}{2}, & \text{si } n \text{ es par} \end{cases}$ | Mediana poblacional $m$ |
| Cuantil muestral de orden $\alpha$ | $\hat{q}_{\alpha} = X_{(k)}, \quad k \approx n\alpha$ | Cuantil poblacional $q_{\alpha}$ |
| Mínimo muestral | $X_{(1)} = \min\{X_1, X_2, \ldots, X_n\}$ | Mínimo poblacional o extremo inferior |
| Máximo muestral | $X_{(n)} = \max\{X_1, X_2, \ldots, X_n\}$ | Máximo poblacional o extremo superior |
| Rango muestral | $R = X_{(n)} - X_{(1)}$ | Amplitud o dispersión poblacional |
| Covarianza muestral | $S_{XY} = \dfrac{1}{n-1}\sum_{i=1}^{n}(X_i-\bar{X})(Y_i-\bar{Y})$ | Covarianza poblacional $\sigma_{XY}$ |
| Coeficiente de correlación muestral | $r = \dfrac{S_{XY}}{S_X S_Y}$ | Correlación poblacional $\rho$ |

## 7.2 Error cuadrático medio y propiedades de un estadístico

### En estadística inferencial, al proponer un estadístico para estimar un parámetro poblacional, no basta con que dicho estadístico tenga una expresión sencilla o que sea fácil de calcular. También es importante evaluar qué tan bueno es como estimador del parámetro de interés. Para ello, se estudian diversas propiedades teóricas que permiten comparar y justificar el uso de un estimador.

### Una de las métricas para evaluar la calidad de un estimador es el **error cuadrático medio**, también llamado **ECM**. Esta cantidad mide, en promedio, qué tan alejado está el estimador $\hat{\theta}$ del verdadero valor del parámetro $\theta$.

### Se define como

### $$ECM(\hat{\theta}) = \mathbb{E}\left[(\hat{\theta}-\theta)^2\right].$$

### Esta expresión penaliza las discrepancias grandes entre $\hat{\theta}$ y $\theta$, ya que el error aparece al cuadrado. Por ello, el ECM no solo toma en cuenta si el estimador tiende a equivocarse, sino también qué tan grandes son esos errores.

### Un estimador con ECM pequeño será, en general, preferible a otro con ECM mayor, pues produce estimaciones más cercanas al parámetro verdadero.

### El ECM puede descomponerse en dos componentes: la varianza del estimador y el cuadrado de su sesgo.

### $$ECM(\hat{\theta}) = \mathrm{Var}(\hat{\theta}) + \left[\mathrm{Sesgo}(\hat{\theta})\right]^2.$$

### Como el sesgo de $\hat{\theta}$ está dado por

### $$\mathrm{Sesgo}(\hat{\theta}) = \mathbb{E}(\hat{\theta}) - \theta,$$

### entonces también puede escribirse

### $$ECM(\hat{\theta}) = \mathrm{Var}(\hat{\theta}) + \left(\mathbb{E}(\hat{\theta})-\theta\right)^2.$$

### Esta descomposición nos indica que un estimador puede tener mal desempeño por dos razones:
### - porque presenta mucha variabilidad de una muestra a otra;
### - porque sistemáticamente sobrestima o subestima al parámetro.

### En consecuencia, al evaluar un estimador se busca un equilibrio entre baja varianza y bajo sesgo.

## Sesgo de un estimador

### El sesgo mide la diferencia entre el valor esperado del estimador y el parámetro que se desea estimar. Si

### $$\mathbb{E}(\hat{\theta}) = \theta,$$

### entonces se dice que $\hat{\theta}$ es un **estimador insesgado** de $\theta$.

### En cambio, si

### $$\mathbb{E}(\hat{\theta}) \neq \theta,$$

### entonces el estimador es **sesgado**.

### Un estimador insesgado no comete error sistemático en promedio. Esto no significa que en cada muestra coincida exactamente con $\theta$, sino que, al considerar muchas repeticiones del muestreo, su promedio coincide con el verdadero valor del parámetro.

## Estimador consistente

### Un estimador $\hat{\theta}_n$ se dice **consistente** para $\theta$ si, cuando el tamaño de muestra $n$ crece, el estimador se aproxima al verdadero valor del parámetro.

### Formalmente, $\hat{\theta}_n$ es consistente si

### $$\hat{\theta}_n \xrightarrow{P} \theta \qquad \text{cuando } n \to \infty,$$

### donde $\xrightarrow{P}$ denota convergencia en probabilidad.

### Intuitivamente, esto significa que, para muestras grandes, la probabilidad de que $\hat{\theta}_n$ esté lejos de $\theta$ se vuelve cada vez más pequeña.

### Una forma frecuente de establecer consistencia es verificar que el error cuadrático medio converge a cero. En efecto, si

### $$ECM(\hat{\theta}_n) \to 0 \qquad \text{cuando } n \to \infty,$$

### entonces $\hat{\theta}_n$ es consistente para $\theta$.

### Esto ocurre porque, si el ECM tiende a cero, entonces tanto la varianza como el sesgo al cuadrado tienden a desaparecer, lo cual implica que el estimador se concentra alrededor del valor verdadero del parámetro.

## Eficiencia

### Cuando se comparan dos estimadores insesgados de un mismo parámetro, suele preferirse aquel que tenga menor varianza.

### Si $\hat{\theta}_1$ y $\hat{\theta}_2$ son insesgados para $\theta$, y además

### $$\mathrm{Var}(\hat{\theta}_1) < \mathrm{Var}(\hat{\theta}_2),$$

### entonces se dice que $\hat{\theta}_1$ es más **eficiente** que $\hat{\theta}_2$.

### En otras palabras, ambos estimadores aciertan en promedio, pero uno presenta menor dispersión muestral y, por tanto, produce estimaciones más estables.

## 7.2.2 Ejemplo

```python
mu=x.mean()
n=20
B=5000

estimador_1=[]
estimador_2=[]
theta_2 = []
theta_3 = []
theta_4 = []
theta_5 = []
mediana = []

for _ in range(B):
    muestra=x.sample(n=n, replace=True)

    mu_est_1=muestra.mean()               # medias
    mu_est_2=muestra.iloc[0]              # el primer valor observado de la muestra
    mu_t_2 = (muestra.iloc[0] + muestra.iloc[1])/2
    mu_t_3 = (muestra.iloc[0] + 2*muestra.iloc[1])/2
    mu_t_4 = min(muestra.iloc[0], muestra.iloc[1], muestra.iloc[2])
    mu_t_5 = sum((muestra.iloc[0], muestra.iloc[1], muestra.iloc[2]))/3
    med = muestra.median()

    estimador_1.append(mu_est_1)
    estimador_2.append(mu_est_2)
    theta_2.append(mu_t_2)
    theta_3.append(mu_t_3)
    theta_4.append(mu_t_4)
    theta_5.append(mu_t_5)
    mediana.append(med)
```

```python
import numpy as np

estimador_1=np.array(estimador_1)
estimador_2=np.array(estimador_2)
theta_2=np.array(theta_2)
theta_3=np.array(theta_3)
theta_4=np.array(theta_4)
theta_5=np.array(theta_5)
mediana = np.array(mediana)

sesgo_1=np.mean(estimador_1)-mu
sesgo_2=np.mean(estimador_2)-mu
sesgo_t_2=np.mean(theta_2)-mu
sesgo_t_3=np.mean(theta_3)-mu
sesgo_t_4=np.mean(theta_4)-mu
sesgo_t_5=np.mean(theta_5)-mu
sesgo_mediana = np.mean(mediana) - mu

var_1=np.var(estimador_1, ddof=1) #insesgado
var_2=np.var(estimador_2, ddof=1)
var_t_2 = np.var(theta_2, ddof=1)
var_t_3 = np.var(theta_3, ddof=1)
var_t_4 = np.var(theta_4, ddof=1)
var_t_5 = np.var(theta_5, ddof=1)
var_mediana = np.var(mediana, ddof=1)

ecm_1=np.mean((estimador_1 - mu)**2)
ecm_2=np.mean((estimador_2 - mu)**2)
ecm_t_2 = np.mean((theta_2 - mu)**2)
ecm_t_3 = np.mean((theta_3 - mu)**2)
ecm_t_4 = np.mean((theta_4 - mu)**2)
ecm_t_5 = np.mean((theta_5 - mu)**2)
ecm_mediana = np.mean((mediana - mu)**2)
```

## 7.3 Máxima Verosimilitud

### La idea intuitiva detrás de este método es que con una muestra de datos $x_1, \dots, x_n$ podamos estimar el parámetro o parámetros, denotados como $\theta$, de tal modo que la probabilidad de obtener la muestra de datos en tal distribución sea máxima.

### Definición: La función de verosimilitud de $n$ variables aleatorias $X_1, X_2, \dots, X_n$ es definida como la función de distribución conjunta de las $n$ variables aleatorias, es decir, $L_{X_1, X_2, \dots, X_n}(x_1, x_2, \dots, x_n; \theta)$ que es considerada una función de $\theta$.
### En particular, si las variables $X_1, X_2, \dots, X_n$ forman una muestra aleatoria, esto es, son independientes e identicamente distribuidas con densidad $f(x;\theta)$, entonces la función de verosimilitud es: $$L_{X_1, X_2, \dots, X_n}(x_1, x_2, \dots, x_n; \theta)=\prod_{i=1}^n f(x_i;\theta)=f(x_1;\theta)\cdot f(x_2;\theta)\cdot\dots\cdot f(x_n;\theta).$$

## 7.3.1 Ejemplo

### Supongamos que tenemos una moneda y queremos estimar la probabilidad $p\in(0, 1)$ de que aparezca cara.
### Podemos modelar la distribución de caras y cruces como una distribución Bernoulli que tiene la siguiente función de masa:
### $$f(x)=p^x (1-p)^{1-x}$$ con $x\in \{0,1\}.$

### Para $n$ lanzamientos independientes de la moneda la función de distribución conjunta será:
### \begin{equation} \begin{split} L_{X_1; X_2, \dots, X_n}(x_1, x_2, \dots, x_n; \theta)& =\prod_{i=1}^n f(x_i;\theta)=f(x_1,\theta)\cdot f(x_2;\theta)\cdot\dots\cdot f(x_n;\theta) \\ &= \prod_{i=1}^n p^{x_i}(1-p)^{1-x_i} \end{split}\end{equation}

```python
from scipy.stats import bernoulli

muestra=bernoulli(0.5).rvs(1000)
```

### Hemos generado una muestra aleatoria de tamaño 1000. Ahora, a través del método de máxima verosimilitud vamos a tratar de estimar el prámetro $p$ con el cual generamos la distribución.  Para ello, primero construyamos a función de verosimilitud $L(\theta)$

```python
import numpy as np
import sympy

x,p=sympy.symbols('x, p')
distr=p**x*(1-p)**(1-x)
L=sympy.prod([distr.subs(x,i) for i in muestra])
L
```

### Construida la función vemos que es una función de una variable, de la cual vamos a encontrar el valor que maximice esta función.

```python
logL=sympy.expand_log(sympy.log(L))
m=sympy.solve(sympy.diff(logL,p),p)
float(*m)
```

## 7.4 Criterios de información: AIC, BIC

### Si varios modelos pueden ajustarse a un mismo conjunto de datos, ¿cómo decidir cuál es más conveniente? Una respuesta importante la proporcionan los **criterios de información**, entre los que destacan el **AIC** y el **BIC**.

### Estos criterios permiten comparar modelos estadísticos combinando:
### - la calidad del ajuste a los datos;
### - la complejidad del modelo.

### La idea general es que un buen modelo debe ajustar razonablemente bien los datos observados, pero sin introducir un número excesivo de parámetros. En otras palabras, se busca un equilibrio entre **bondad de ajuste** y **parsimonia**.

### Supóngase que se tiene una muestra aleatoria $X_1, X_2, \ldots, X_n$ y que se propone un modelo paramétrico con función de verosimilitud

### $$L(\theta)=L(x_1,x_2,\ldots,x_n; \theta).$$

### Si $\hat{\theta}$ es el estimador de máxima verosimilitud, entonces el valor $L(\hat{\theta})$ representa la máxima verosimilitud alcanzada por el modelo con los datos observados. En particular con

### $$\ell(\hat{\theta})=\log L(\hat{\theta}).$$

### Cuanto mayor sea este valor, mejor ajuste ofrece el modelo a los datos. Sin embargo, un modelo con más parámetros casi siempre puede ajustarse mejor, por lo que no conviene juzgar únicamente con base en la verosimilitud máxima.

### El **AIC** se define por

### $$AIC = -2\log L(\hat{\theta}) + 2k,$$

### donde:
### - $L(\hat{\theta})$ es la verosimilitud evaluada en el estimador de máxima verosimilitud;
### - $k$ es el número de parámetros del modelo.

### Si se comparan varios modelos ajustados a los mismos datos, se prefiere aquel que tenga el **menor valor de AIC**. En términos generales, el AIC se asocia con la capacidad predictiva del modelo, pues busca seleccionar aquel que perdería menos información al representar el mecanismo generador de los datos.
### El **BIC** se define por

### $$BIC = -2\log L(\hat{\theta}) + k\log n,$$

### donde:
### - $L(\hat{\theta})$ es la verosimilitud máxima;
### - $k$ es el número de parámetros;
### - $n$ es el tamaño de muestra.

### Al igual que el AIC, el primer término evalúa el ajuste del modelo, mientras que el segundo penaliza su complejidad. La diferencia es que en el BIC la penalización depende del tamaño de muestra, pues aparece el factor $\log n$. Al comparar varios modelos sobre la misma muestra, se prefiere aquel que tenga el **menor valor de BIC**.
### En conclusión:
### - ambos premian modelos con mayor verosimilitud;
### - ambos penalizan el número de parámetros;
### - el BIC suele ser más estricto que el AIC cuando $n$ es grande.

### Por lotanto, no siempre ambos criterios seleccionan el mismo modelo. El AIC tiende a ser más flexible con modelos relativamente complejos, mientras que el BIC suele inclinarse por modelos más parsimoniosos.


| Situación | Criterio que suele preferirse |
|---|---|
| Interesa principalmente predecir | $AIC$ |
| Interesa un modelo más simple e interpretable | $BIC$ |
| Se tolera algo más de complejidad si mejora el ajuste | $AIC$ |
| Se desea penalizar fuertemente el número de parámetros | $BIC$ |
| Muestra grande y búsqueda de parsimonia | $BIC$ |

```python
def calcular_bic_aic(datos, distribucion, parametros):
    n=len(datos)
    k=len(parametros)
    log_likelihood=np.sum(distribucion.logpdf(datos, *parametros))

    aic=-2 * log_likelihood + 2 * k
    bic=-2 * log_likelihood + k * np.log(n)

    return bic, aic
```

```python
def ajustar_distribucion(datos, distribuciones):
    resultados=[]

    for distribucion in distribuciones:
        parametros=distribucion.fit(datos)
        bic, aic=calcular_bic_aic(datos, distribucion, parametros)
        resultados.append({
                "Distribución": distribucion.name,
                "Parámetros": parametros,
                "AIC": aic,
                "BIC": bic
            })

    tabla=pd.DataFrame(resultados).sort_values("BIC").reset_index(drop=True)

    return tabla
```

```python
from scipy.stats import gamma, lognorm, beta, chi2

distribuciones=[norm, expon, gamma, beta, chi2]
datos_ajuste=norm.rvs(loc=0, scale=1, size=1000)
ajustar_distribucion(datos_ajuste, distribuciones)
```
