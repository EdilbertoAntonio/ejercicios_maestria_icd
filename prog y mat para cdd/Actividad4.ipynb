```python
Problema = [1, 2, 3, 4, 5]
Grupo_A = [28,29,30,31,32]
Grupo_B = [10,20,30,40,50]
```

Considero que ambos grupos presentan desempeños consistentes debido a que los problemas fueron aumentando, los minutos para resolverlos fueron aumentando y el tiempo que hay entre cada minuto que tardaron en resolver los problemas es el mismo para cada grupo, es decir, los minutos de resolución aumentan proporcionalmente.

2. El grupo que tendran mayor varianza en el grupo B, debido a que el rango de sus datos se encuentra más disperso.

3. De igual manera, el grupo con mayor desviación estándar será el grupo B, debido a que se relaciona directamente con la varianza.

## Parte 2

```python
# calculado manualmente

def media(datos:list)->float:
    N = len(datos)
    suma = 0
    for i in range(N):
        suma = suma + datos[i]
    return suma/N

def varianza_poblacional(datos:list, media:float)->float:
    N = len(datos)
    suma = 0
    for i in range(N):
        suma = suma + (datos[i] - (media))**2
    return suma/N

def listas(datos:list, media:float)->list:
    N = len(datos)
    # suma = 0
    dif = []
    dif_2 = []
    for i in range(N):
        diferencia = datos[i] - (media)
        dif.append(diferencia)
        dif_2.append(diferencia**2)
    dif.append(sum(dif))
    dif_2.append(sum(dif_2))
    return [dif, dif_2]

media_grupo_A = media(Grupo_A)
varianzap_grupo_A = varianza_poblacional(Grupo_A, media_grupo_A)
des_estandarp_grupo_A = (varianzap_grupo_A)**(1/2)

media_grupo_B = media(Grupo_B)
varianzap_grupo_B = varianza_poblacional(Grupo_B, media_grupo_B)
des_estandarp_grupo_B = (varianzap_grupo_B)**(1/2)

diferencias_A = listas(Grupo_A, media_grupo_A)
Grupo_A.append('Suma')
print('\nGrupo A')
print(f'{'x_i':<10} {'x_i - x_prom':<15} {'(x_i - x_prom)^2':<10}')
for i in range(6):
    print(f'{Grupo_A[i]:<10} {diferencias_A[0][i]:<15} {diferencias_A[1][i]:<10}')
Grupo_A.pop()

diferencias_B = listas(Grupo_B, media_grupo_B)
Grupo_B.append('Suma')
print('\nGrupo B')
print(f'{'x_i':<10} {'x_i - x_prom':<15} {'(x_i - x_prom)^2':<10}')
for i in range(6):
    print(f'{Grupo_B[i]:<10} {diferencias_B[0][i]:<15} {diferencias_B[1][i]:<10}')
Grupo_B.pop()

print(f'\n{'Medida':<10} {'Grupo A':<10} {'Grupo B':<10}')
print(f'{'Media':<10} {media_grupo_A:<10} {media_grupo_B:<10}')
print(f'{'Varianza':<10} {varianzap_grupo_A:<10} {varianzap_grupo_B:<10}')
print(f'{'Desv. Est':<10} {des_estandarp_grupo_A:<10.2f} {des_estandarp_grupo_B:<10.2f} ')
```

## Parte 3

1. Significa que el rango en el que se mueven los datos es corto, por lo tanto la concentración de los datos esta bastante junta.

2. Una desviación estándar grande significa que los datos están muy separados y por lo tanto esten más dispersos, por lo que la concentración de los datos no es alta.

3. Ambos tienen la misma media, pero al contrario sus varianzas y desviaciones estándar son diferentes, en el grupo A en más pequeña que el grupo B, por lo tanto la concentración de los datos del grupo A es mucho mayor que la del grupo B

## Parte 4

```python
# Con la libreria de statistics

import statistics as st

media_grupo_A = st.mean(Grupo_A)
varianzap_grupo_A = st.pvariance(Grupo_A)
des_estandarp_grupo_A = st.pstdev(Grupo_A)

media_grupo_B = st.mean(Grupo_B)
varianzap_grupo_B = st.pvariance(Grupo_B)
des_estandarp_grupo_B = st.pstdev(Grupo_B)

print(f'{'Medida':<10} {'Grupo A':<10} {'Grupo B':<10}')
print(f'{'Media':<10} {media_grupo_A:<10} {media_grupo_B:<10}')
print(f'{'Varianza':<10} {varianzap_grupo_A:<10} {varianzap_grupo_B:<10}')
print(f'{'Desv. Est':<10} {des_estandarp_grupo_A:<10.2f} {des_estandarp_grupo_B:<10.2f} ')
```

Ambos resultados fueron los mismos, por lo tanto la forma en que los calculé estuvo bien, sin embargo un ligera diferencia fue que con el modulo 'statistics' me devolvió resultados de tipo 'int' mientras que mis calculos están en tipo 'float'
