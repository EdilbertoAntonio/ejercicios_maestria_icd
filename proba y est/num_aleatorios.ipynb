```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

semilla = 7
Z = [semilla]

for i in range(20):
    aux = (5*Z[i]+3) % 16
    Z.append(aux)

U = [None]+[i/16 for i in Z[1:]]

tabla = pd.DataFrame({'Z_i':Z, 'U_i':U})
tabla
```

```python
N = 10000
S_x = np.random.uniform(-1,1,N)
S_y = np.random.uniform(-1,1,N)

def dentro(x, y):
    return np.where((x**2 + y**2)<=1, 'adentro', 'afuera')

estado = dentro(S_x, S_y)

puntos = pd.DataFrame({'X':S_x, 'Y': S_y, 'Estado': estado})
puntos.head()

plt.figure(figsize=(10, 10))
sns.scatterplot(x='X', y='Y', hue='Estado', data=puntos, 
                palette={'adentro': '#E61A33', 'afuera': '#2683C6'})
plt.show()

n_adentro = len(puntos[puntos['Estado'] == 'adentro'])
estimacion_pi = 4 * n_adentro/N
print(estimacion_pi)
# plt.scatter(S_x, S_y)
```

```python
N = 50000000
S_x = np.random.uniform(-1, 1, N)
S_y = np.random.uniform(-1, 1, N)

n_adentro = np.sum((S_x**2 + S_y**2) <= 1)

estimacion_pi = 4 * n_adentro / N
print(estimacion_pi)
```

```python
# generador de Park-Miller
import numpy as np
import pandas as pd

semilla = 166791790
Z = [semilla]
m = 2**31-1
a = 16807
n = 10

for i in range(n):
    aux = (a*Z[i]) % m
    Z.append(aux)

U = [z/m for z in Z[1:]]

tabla = pd.DataFrame({'Z_i':Z, 'U_i':[np.nan]+U})
tabla
```
