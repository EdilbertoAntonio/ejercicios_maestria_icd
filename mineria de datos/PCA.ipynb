<h1>PCA Analysis</h1>
<div style="text-align:justify">Principal Component Analysis (PCA) serves as a linear dimensionality reduction technique utilized for exploratory data analysis, visualization, and data preprocessing. Through linear transformation, the data is projected onto a new coordinate system, revealing principal components—unit vectors that indicate directions capturing the maximum data variation.<br>
In a real coordinate space, the principal components are a sequence of vectors. Each vector represents the direction of a line that optimally fits the data while remaining orthogonal to the preceding vectors. "best-fitting line" refers to one minimizing the average squared perpendicular distance from the data points to the line. These principal components form an orthonormal basis, ensuring that individual dimensions of the data are linearly uncorrelated.<br>
PCA is commonly employed for visualization purposes in data science for the following reasons:<br>
*Reduce Dimensionality:* <br>
In high-dimensional datasets, it can be challenging to visualize and interpret the relationships between variables. PCA reduces the dimensionality of the data by projecting it onto a lower- dimensional subspace while retaining as much variance as possible.<br>
*Identify Important Features:* <br>
PCA helps identify the most significant features or variables in the dataset. The first few principal components often capture the majority of the variability, allowing for a more focused analysis of the essential features.<br>
*Visualize Patterns and Clusters:* <br>
By plotting the data in the reduced-dimensional space of principal components, patterns, clusters, and trends that may not be apparent in the original high-dimensional space can become more visible. This aids in the exploration and understanding of the structure within the data.<br>
*Remove Redundancy:* <br>
PCA removes correlations between variables, resulting in uncorrelated principal components. This can be beneficial when dealing with multicollinearity or redundant information in the dataset.
</div>

<h1>PCA Analysis</h1>
<div style="text-align:justify"> 
As the number of dimensions increases, the number of possible combinations of features increases exponentially, which makes it computationally difficult to obtain a representative sample of the data and it becomes expensive to perform tasks such as clustering or classification. To address the curse of dimensionality, Feature engineering techniques are used which include feature selection and feature extraction. Principal Component Analysis(PCA) is dimensionality reduction technique that aims to reduce the number of input features while retaining as much of the original information as possible.
Principal Component Analysis (PCA) is a statistical procedure that uses an orthogonal transformation that converts a set of correlated variables to a set of uncorrelated variables.
</div>

- Principal Component Analysis (PCA) is a method used to reduce the dimensionality of data by identifying a set of perpendicular axes known as principal components. These components, which are linear combinations of the original variables, are arranged in descending order based on their significance in capturing the maximum variance within the dataset. The cumulative variance explained by all principal components is equivalent to the total variance present in the original dataset.
- The initial principal component captures the greatest variability within the data, while the second principal component captures the highest variance perpendicular to the first principal component, and this pattern continues for subsequent components
- In Principal Component Analysis, it is assumed that the information is carried in the variance of the features, that is, the higher the variation in a feature, the more information that features carries

>**Covariance Matrix Computation**<br>
Covariance measures the strength of joint variability between two or more variables, indicating how much they change in relation to each other. To find the covariance we can use the formula:
$cov(x_1,x_2) = \frac{\sum_{i=1}^{n}(x_{1i}-\bar{x_1})(x_{2i}-\bar{x_2})}{n-1}$<br>
The value of covariance can be positive, negative, or zeros

>**Compute Eigenvalues and Eigenvectors of Covariance Matrix**<br>
Let A be a square $n\times n$ matrix and X be a non-zero vector for which 
$AX = \lambda X$
for some scalar values $\lambda$.<br>
Using,
$|A-\lambda I|=0$<br>
we can find the eigenvalues $\lambda$, and therefore corresponding eigenvector can be found using the equation $AX = \lambda X$.

PCA employs a linear transformation that is based on preserving the most variance in the data using the least number of dimensions.

### Estandarizamos

```python
#compute mean and SD
x= df[cancer['feature_names']]
# Mean
x_mean = x.mean()

# Standard deviation
x_std = x.std(ddof=0)

# Standardization
Z = (x- x_mean) / x_std

Z.info()
```

```python
#import matplotlib.pyplot as plt
import seaborn as sns
#Computing covariance matrix
c = Z.cov()
# Plot the covariance matrix
sns.heatmap(c)
plt.show()
```

### eigenvectors y eigenvalues

```python
#Computing eigenvectors and eigenvalues
import numpy as np
eigenvalues, eigenvectors = np.linalg.eig(c)
```

```python
eigenvalues
```

### ordenamiento

```python
#Sorting the eigenvalues

idx = eigenvalues.argsort()[::-1] # mayor a menor pero es el indice del mayor a menor
eigenvalues = eigenvalues[idx]
eigenvectors = eigenvectors[:,idx]
idx
```

**Explained variance** is a statistical measure of how much variation in a dataset can be attributed to each of the principal components (eigenvectors) generated by the principal component analysis (PCA) method. In other words, it tells us how much of the total variance is “explained” by each component. This is important because it allows us to rank the components in order of importance, and to focus on the most important ones when interpreting the results of our analysis. 

In general, the larger the variance explained by a principal component, the more important that component is. Explained variance can be used to choose the number of dimensions to keep in a reduced dataset.

```python
explained_var = np.cumsum(eigenvalues) / np.sum(eigenvalues) #como aumenta la varianza conforme a los compoenentes
explained_var
```

```python
plt.plot(explained_var, marker='o')
```

The choice of the percentage of explained variance to retain in Principal Component Analysis (PCA) depends on the specific goals of your analysis, the characteristics of your data, and the trade-off between dimensionality reduction and retaining information.

**70-80% Explained Variance:**
Retaining 70-80% of the explained variance is often considered a moderate level of dimensionality reduction. This range may strike a balance between reducing dimensionality and retaining enough information for many applications.

**90% Explained Variance:**
Choosing to retain 90% of the explained variance is a more conservative approach. This level of retention is appropriate when a higher level of information preservation is desired, and sacrificing some dimensionality reduction is acceptable.

**95% Explained Variance or Higher:**
Retaining 95% or more of the explained variance is a conservative choice that preserves a substantial amount of information. This is typically preferred in applications where maintaining a high level of fidelity to the original data is crucial.

**Scree Plot Analysis:**<br>
Analyzing a scree plot or cumulative explained variance plot can guide the decision. Look for an "elbow" in the plot or a point where adding more components doesn't significantly increase the explained variance. This can help in determining a suitable cutoff.

```python
n_components = np.argmax(explained_var >= 0.50) + 1
n_components
```

**Project the Data onto the Selected Principal Components**<br>
Find the projection matrix, It is a matrix of eigenvectors corresponding to the largest eigenvalues of the covariance matrix of the data. it projects the high-dimensional dataset onto a lower-dimensional subspace
The eigenvectors of the covariance matrix of the data are referred to as the principal axes of the data, and the projection of the data instances onto these principal axes are called the principal components. 

```python
# PCA component or unit matrix
import pandas as pd
u = eigenvectors[:,:n_components]
pca_component = pd.DataFrame(u,index = cancer['feature_names'],columns = ['PC1','PC2'])
```

```python
pca_component
```

```python
#heatmap
plt.figure(figsize =(5, 7))
sns.heatmap(pca_component)
plt.title('PCA Component')
plt.show()
```

Then, we project our dataset using the formula:  
$\begin{aligned} Proj_{P_i}(u) &= \frac{P_i\cdot u}{|u|} \\ &=P_i\cdot u \end{aligned}$       

Dimensionality reduction is then obtained by only retaining those axes (dimensions) that account for most of the variance, and discarding all others.



The principal axes of the data are the eigenvectors of its covariance matrix, and the corresponding projections of data instances onto these axes are known as the principal components. Dimensionality reduction involves retaining only those axes (dimensions) that capture the majority of the variance while discarding the rest.

<h2>PCA using Using Sklearn</h2>
There are different libraries in which the whole process of the principal component analysis has been automated by implementing it in a package as a function and we just have to pass the number of principal components which we would like to have. Sklearn is one such library that can be used for the PCA

```python
# Import PCA
from sklearn.decomposition import PCA

# Choosing 2 components
pca = PCA(n_components=2)
pca.fit(Z)
x_pca=pca.transform(Z)
 
# Create the dataframe
df_pca1 = pd.DataFrame(x_pca,columns=['PC{}'.format(i+1) for i in range(n_components)])
print(df_pca1)
```

```python
plt.figure(figsize=(8, 6))
 
plt.scatter(df_pca1.iloc[:,0], df_pca1.iloc[:,1],c=cancer['target'],cmap='plasma',s=2)
 
# labeling x and y axes
plt.xlabel('First Principal Component')
plt.ylabel('Second Principal Component')
plt.show()
```

<h2>Advantages of Principal Component Analysis</h2>
**Dimensionality Reduction:** By reducing the number of variables, PCA simplifies data analysis, improves performance, and makes it easier to visualize data.<br>
**Feature Selection:** Principal Component Analysis can be used for feature selection, which is the process of selecting the most important variables in a dataset. This is useful in machine learning, where the number of variables can be very large, and it is difficult to identify the most important variables.<br>
**Data Visualization:** Principal Component Analysis can be used for data visualization. By reducing the number of variables, PCA can plot high-dimensional data in two or three dimensions, making it easier to interpret.

<h2>Disadvantages of Principal Component Analysis</h2>
Interpretation of Principal Components: The principal components created by Principal Component Analysis are linear combinations of the original variables, and it is often difficult to interpret them in terms of the original variables. This can make it difficult to explain the results of PCA to others.<br>
Data Scaling: Principal Component Analysis is sensitive to the scale of the data. If the data is not properly scaled, then PCA may not work well. Therefore, it is important to scale the data before applying Principal Component Analysis.<br>
Information Loss: Principal Component Analysis can result in information loss. While Principal Component Analysis reduces the number of variables, it can also lead to loss of information. The degree of information loss depends on the number of principal components selected. Therefore, it is important to carefully select the number of principal components to retain.
