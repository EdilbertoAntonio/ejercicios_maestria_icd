```python
# 5.2

def mystery(sequence):
    return sequence == sorted(sequence)

# devuelve true si la lista que le pases está ordenada, false en caso contrario

lista_ejemplo = [15, 1, 47, 2, 12]
print(mystery(lista_ejemplo))
lista_ordenada = [1,2,3,4,5]
print(mystery(lista_ordenada))
```

```python
# 5.7

def uniques(lista):
    unicos = []
    for elemento in lista:
        if lista.count(elemento) == 1:
            unicos.append(elemento)
        else:
            if elemento not in unicos:
                unicos.append(elemento)

    return sorted(unicos)

prueba = [2,2,4,5,6,7,6,5,1,8,1,8,1,1,3]
uniques(prueba)
```
