```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
from scipy.ndimage import gaussian_filter, uniform_filter, median_filter
```

```python
rng = np.random.default_rng(seed=42)

image = Image.open('datos\\Dog.jpg').convert('L')
original = np.asarray(image,dtype=np.float64) # matrix to be used for smoothing
noise = rng.normal(loc = 0, scale = 150, size = original.shape) # add noise to the original image

distorted = original + noise

# apply a function such that, 0 < I_ij < 255 If less than a zero it assigns a zero val
# If greater tahn 255 it assigns 255

# distorted = np.where(distorted > 255, 255, distorted)
# distorted = np.where(distorted < 0, 0, distorted)

distorted = np.clip(distorted, 0, 255)

# smoothingguassian_filter by mean, gaussian and median

mean_filter = uniform_filter(distorted, size = 8)
median = median_filter(distorted, size = 10)
gaussiano = gaussian_filter(distorted, sigma = 0.1, mode='reflect') 


fig, ax = plt.subplots(1,5,figsize=(15,8))
ax[0].imshow(original, cmap ='gray', vmin=0, vmax=255)
ax[0].set_title('original')
ax[1].imshow(distorted, cmap ='gray', vmin=0, vmax=255)
ax[1].set_title('distorted')
ax[2].imshow(mean_filter, cmap ='gray', vmin=0, vmax=255)
ax[2].set_title('mean filter')
ax[3].imshow(median, cmap ='gray', vmin=0, vmax=255)
ax[3].set_title('median filter')
ax[4].imshow(gaussiano, cmap ='gray', vmin=0, vmax=255)
ax[4].set_title('gaussian filter')
ax[4].set_yticks([])
ax[3].set_yticks([])
ax[2].set_yticks([])
ax[1].set_yticks([])
ax[0].set_yticks([])

```
