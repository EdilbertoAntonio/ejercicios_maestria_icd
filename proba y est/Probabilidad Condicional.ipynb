## 2 Probabilidad Condicional

### 2.1 Definición

### Dado un espacio de probabilidad $(\Omega, \mathcal{F}, \mathbb{P})$ y sea $B\in \mathcal{F}$ con $\mathbb{P}(B)>0$. Definimos la probabilidad condicional dado el evento $B$, como $$\mathbb{P}(A|B)=\frac{\mathbb{P}(A\cap B)}{\mathbb{P}(B)}.$$

### 2.1.1 Ejemplo

```python
import pandas as pd
import numpy as np

data=pd.DataFrame({'genero': np.repeat(np.array(['Masculino', 'Femenino']), 150),
                   'deporte': np.repeat(np.array(['Béisbol', 'Basket', 'Americano','Soccer', 
                                                  'Béisbol', 'Basket', 'Americano', 'Soccer']), 
                                                   (34, 40, 58, 18, 34, 52, 20, 44))})

data.head()
```

```python
data.tail()
```

### Una tabla de contingencia es una herramienta utilizada en la rama de la estadística, la cual consiste en crear al menos dos filas y dos columnas para representar datos categóricos en términos de conteos de frecuencia. Esta herramienta, que también se conoce como tabla cruzada o como tabla de dos vías, tiene el objetivo de representar en un resumen, la relación entre diferentes variables categóricas. 

```python
tabla = pd.crosstab(index=data['genero'], columns=data['deporte'], margins=True)
tabla
```

### Calcular la probabilidad de seleccionar a un individuo de la población encuestada del genero másculino dado que prefiere el béisbol. 

Evento B (es el supuesto) que le guste beisbol, A que sea hombre

```python
tabla.loc['Masculino', 'Béisbol']/tabla.loc['All', 'Béisbol'] 
```

### Calcular la probabilidad de que prefiera basket dado que el individuo es del genero femenino. 

B supuesto, que sea femenino, A que le guste basket

```python
tabla.loc['Femenino', 'Basket']/tabla.loc['Femenino', 'All']
```

## 2.2 Regla de Probabilidad Condicional

### Las siguientes reglas son de gran ayuda en el cálculo de probabilidades condicionales:
### 1. Regla del producto $$\mathbb{P}(A_1\cap A_2\cap...\cap A_n)=\mathbb{P}(A_n|A_1\cap \ldots \cap A_{n-1})\mathbb{P}(A_{n-1}|A_1\cap \ldots \cap A_{n-2})\ldots \mathbb{P}(A_2|A_1)\mathbb{P}(A_1)$$

### 2. Regla de la Probabilidad Total $$\mathbb{P}(B)=\sum_{i=1}^{n}\mathbb{P}(B|A_i)\mathbb{P}(A_i)$$

### 3. Regla de Bayes $$\mathbb{P}(A_i|B)=\frac{\mathbb{P}(B|A_i)\mathbb{P}(A_i)}{\sum_{i=1}^{n}\mathbb{P}(B|A_i)\mathbb{P}(A_i)}$$

La reglas 2 y 3 están relacionadas, de la 2 se puede sacar la 3

## Ejemplo 2.2.1

### Tres máquinas de cierta planta de ensamble, $A_1,A_2, A_3,$ montan $30\%, 45\%, 25\%$ de los productos, respectivamente. Se sabe por experiencia que $2\%,3\%, 2\%$ de los productos ensamblados por cada máquina, respectivamente, tienen defectos. Ahora bien, supóngase que se selecciona de forma aleatoria un producto terminado. ¿Cuál es la probabilidad de que esté defectuoso?

```python
P_A = [0.30, 0.45, 0.25]
P_D_dado_A = [0.02, 0.03, 0.02]
```

```python
sum(P_A[i] * P_D_dado_A[i] for i in range(len(P_A))) #la regla de la probabilidad total, 2
```

### si se elige al azar un producto y se encuentra que es defectuoso ¿cuál es la probabilidad de que este ensamblado por la máquina $A_3$?

```python
(P_A[2] * P_D_dado_A[2]) / (sum(P_A[i] * P_D_dado_A[i] for i in range(len(P_A)))) #regla de bayes, la 3
#una forma de actualizar la probabilidad
```

## Ejemplo 2.2.3


### Considérese una base de datos de pasajeros del Titanic. Sean los eventos:

### $A$: “el pasajero sobrevivió”;
### $B$: “el pasajero es mujer y viaja en primera clase”.

### La probabilidad condicional $\mathbb P(A\mid B)$ puede estimarse mediante la frecuencia relativa dentro de la base de datos, es decir, restringiendo el análisis únicamente a los registros que cumplen la condición $B$.

```python
import seaborn as sns

data_titanic=sns.load_dataset('titanic')
data_titanic.head()
```

```python
casos = data_titanic[(data_titanic['sex'] == 'female') & (data_titanic['class'] == 'First')]
casos.head()
```

```python
sum(casos['survived'])/len(casos['survived'])
```

```python
casos2 = data_titanic[(data_titanic['sex'] == 'male') & (data_titanic['class'] == 'Third')]
casos2.head()
```

```python
sum(casos2['survived'])/len(casos2['survived'])
```
