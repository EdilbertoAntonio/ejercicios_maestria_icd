## 6.7

Recall that strings are sequences of characters. Use techniques similar to Fig. 6.2 to write a script that inputs a sentence from the user, then uses a dictionary to summarize the number of occurrences of each letter. Ignore case, ignore blanks and assume the user does not enter any punctuation. Display a two-column table of the letters and their counts with the letters in sorted order. Challenge: Use a set operation to determine which letters of the alphabet were not in the original string.

```python
entrada_usuario = str(input('Escribe tu texto:'))
entrada_usuario = entrada_usuario.lower().strip().replace(' ','')

conjunto_alfabeto = set("abcdefghijklmnopqrstuvwxyz")
conjunto_entrada = set(entrada_usuario)
conjunto_faltante = conjunto_alfabeto - conjunto_entrada

conteo_letras = {}
for letra in entrada_usuario:
    if letra in conteo_letras:
        conteo_letras[letra] += 1 
    else:
        conteo_letras[letra] = 1

print(f'{"WORD":<12}COUNT')
for letra, count in sorted(conteo_letras.items()):
    print(f'{letra:<12}{count}')

print(f'Letras faltantes del abecedario:')
print(sorted(conjunto_faltante))
```

## 6.8

In check-writing systems, it’s crucial to prevent alteration of check amounts. One common security method requires that the amount be written in numbers and spelled out in words as well. Even if someone can alter the numerical amount of the check, it’s tough to change the amount in words. Create a dictionary that maps numbers to their corresponding word equivalents. Write a script that inputs a numeric check amount that’s less than 1000 and uses the dictionary to write the word equivalent of the amount. For example, the amount 112.43 should be written as:

ONE HUNDRED TWELVE AND 43/100

```python
mapa_numeros = {
    'unidades': {1: 'UNO', 2: 'DOS', 3: 'TRES', 4: 'CUATRO', 5: 'CINCO', 6: 'SEIS', 7: 'SIETE', 8: 'OCHO', 9: 'NUEVE'},
    
    'especiales': {10: 'DIEZ', 11: 'ONCE', 12: 'DOCE', 13: 'TRECE', 14: 'CATORCE', 15: 'QUINCE', 
                   16: 'DIECISEIS', 17: 'DIECISIETE', 18: 'DIECIOCHO', 19: 'DIECINUEVE', 
                   20: 'VEINTE', 21: 'VEINTIUNO', 22: 'VEINTIDOS', 23: 'VEINTITRES', 24: 'VEINTICUATRO', 
                   25: 'VEINTICINCO', 26: 'VEINTISEIS', 27: 'VEINTISIETE', 28: 'VEINTIOCHO', 29: 'VEINTINUEVE'},
    
    'decenas': {3: 'TREINTA', 4: 'CUARENTA', 5: 'CINCUENTA', 6: 'SESENTA', 7: 'SETENTA', 8: 'OCHENTA', 9: 'NOVENTA'},
    
    'centenas': {1: 'CIENTO', 2: 'DOSCIENTOS', 3: 'TRESCIENTOS', 4: 'CUATROCIENTOS', 5: 'QUINIENTOS', 
                 6: 'SEISCIENTOS', 7: 'SETECIENTOS', 8: 'OCHOCIENTOS', 9: 'NOVECIENTOS'}
}

def convertir_cheque_espanol(cantidad_str):
    if '.' in cantidad_str:
        entero_str, centavos = cantidad_str.split('.')
        centavos = centavos.ljust(2, '0')[:2] 
    else:
        entero_str = cantidad_str
        centavos = '00'
        
    numero = int(entero_str)
    palabras = []

    if numero == 0:
        return f"CERO CON {centavos}/100"
    if numero == 100:
        return f"CIEN CON {centavos}/100"

    centena_digito = numero // 100
    if centena_digito > 0:
        palabras.append(mapa_numeros['centenas'][centena_digito])

    resto = numero % 100
    if resto > 0:
        if resto in mapa_numeros['especiales']:
            palabras.append(mapa_numeros['especiales'][resto])
        else:
            decena_digito = resto // 10
            unidad_digito = resto % 10
            
            if decena_digito > 0:
                palabras.append(mapa_numeros['decenas'][decena_digito])
            
            if unidad_digito > 0:
                palabras.append(f"Y {mapa_numeros['unidades'][unidad_digito]}")

    letras = " ".join(palabras)
    return f"{letras} CON {centavos}/100"

entrada = input("Ingresa la cantidad del cheque (menor a 1000): ")
resultado = convertir_cheque_espanol(entrada)
print(f"La cantidad ${entrada} en el cheque se escribirá así:{resultado}")
```

## 6.10

Using the following sets:

{'red', 'green', 'blue'}

{'cyan', 'green', 'blue', 'magenta', 'red'}

display the results of:

a) comparing the sets using the each of the comparison operators.

b) combining the sets using each of the mathematical set operators.

```python
setA = {'red', 'green', 'blue'}
setB = {'cyan', 'green', 'blue', 'magenta', 'red'}

# a)
igualdad = (setA == setB)
print(f'Los conjunto son iguales: {igualdad}')
desigualdad = (setA != setB)
print(f'Los conjunto son desiguales: {desigualdad}')
subconjunto = (setA <= setB)
print(f'El conjunto A es subconjunto de B: {subconjunto}')
subconjunto_propio = (setA < setB)
print(f'El conjunto A es subconjunto propio de B: {subconjunto_propio}')
superconjunto = (setA >= setB)
print(f'El conjunto A es superconjunto de B: {superconjunto}')
superconjunto_propio = (setA > setB)
print(f'El conjunto A es superconjunto propio de B: {superconjunto_propio}')

# b)
union = setA | setB
print(f'union de A con B es {union}')
interseccion = setA & setB
print(f'interseccion de A con B es {interseccion}')
diferencia = setA - setB
print(f'diferencia de A con B es {diferencia}')
diferencia_simetrica = setA ^ setB
print(f'diferencia simetrica de A con B es {diferencia_simetrica}')

```

## 6.11

Modify the script of Fig. 4.2 to play 1,000,000 games of craps. Use a wins dictionary to keep track of the number of games won for a particular number of rolls. Similarly, use a losses dictionary to keep track of the number of games lost for a particular number of rolls. As the simulation proceeds, keep updating the dictionaries.

A typical key–value pair in the wins dictionary might be 4: 50217 indicating that 50217 games were won on the 4th roll. Display a summary of the results including:

a) the percentage of the total games played that were won.

b) the percentage of the total games played that were lost.

c) the percentages of the total games played that were won or lost on a given roll (column 2 of the sample output).

d) the cumulative percentage of the total games played that were won or lost up to and including a given number of rolls (column 3 of the sample output).

```python
import random

def roll_dice():
    die1 = random.randrange(1, 7)
    die2 = random.randrange(1, 7)
    return (die1, die2) 

simulaciones = 1000000
wins = {}
loses = {}

for _ in range(simulaciones):
    die_values = roll_dice() 

    sum_of_dice = sum(die_values)

    if sum_of_dice in (7, 11): 
        game_status = 'WON'
    elif sum_of_dice in (2, 3, 12): 
        game_status = 'LOST'
    else: 
        game_status = 'CONTINUE'
        my_point = sum_of_dice

    tiro = 1
    while game_status == 'CONTINUE':
        die_values = roll_dice()
        sum_of_dice = sum(die_values)

        if sum_of_dice == my_point: 
            game_status = 'WON'
        elif sum_of_dice == 7: 
            game_status = 'LOST'

        tiro += 1

    if game_status == 'WON':
        wins[tiro] = wins.get(tiro, 0) + 1
    else:
        loses[tiro] = loses.get(tiro, 0) + 1
```

```python
# a)

porcentaje_ganados = (sum(wins.values())/simulaciones) * 100
print(f'el porcentaje de juegos ganados fue: {porcentaje_ganados:.2f}%')

# b)

porcentaje_perdidos = (sum(loses.values()) /simulaciones) * 100
print(f'el porcentaje de juegos ganados fue: {porcentaje_perdidos:.2f}%')

# c) y d)

tiros_ganados = {tiro: ganados/simulaciones for tiro, ganados in wins.items()}
tiros_perdidos = {tiro: perdidos/simulaciones for tiro, perdidos in loses.items()}

tiro_maximo = max(max(wins.keys()), max(loses.keys()))

cumulativo = 0
print('porcentaje de ganados/perdidos basado en los tiros\n')
print(f"{'Tiros':<10} {'Porcentaje':<10} {'Acumulado':<10}")
for i in range(1,tiro_maximo+1):
    total_games = tiros_ganados.get(i, 0) + tiros_perdidos.get(i, 0)
    cumulativo += total_games
    print(f"{i:<10} {total_games:<10.4%} {cumulativo:<10.4%}")
```
