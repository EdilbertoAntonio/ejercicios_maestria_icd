```python
import os
from dotenv import load_dotenv
import requests
import json
```

```python
load_dotenv() # nos permite cargar la llave de la api
token = os.getenv("INEGI_API_KEY")

IdIndicador = '1002000001'
Idioma = 'es'
# Puede ser nacional [00], por entidad federativa [99] o por municipio [999], dependiendo de cada indicador.
Area_Geografica = '00'
# Puede consultarse solo el dato más reciente [true] o la serie histórica completa [false].
Recientes = 'true'
# Corresponde a la fuente de diseminación de donde se obtendrán los datos consultados.
Fuente_datos = 'BISE'
# Con él se identificará la edición [2.0] del servicio de provisión de datos.
Version = '2.0'
# Se ofrece la información en 3 tipos de formatos: JSON [json], JSONP [jsonp] o XML [xml].
Formato = 'json'

url = f'https://inegi.org.mx/app/api/indicadores/desarrolladores/jsonxml/INDICATOR/{IdIndicador}/{Idioma}/{Area_Geografica}/{Recientes}/{Fuente_datos}/{Version}/{token}?type={Formato}'

response = requests.get(url)
if response.status_code == 200:
    data = response.json()
    print(data)
else:
    print('Error en la consulta')
```

```python
def consultar_poblacion(Area_Geografica):
    # Construimos la URL con la región específica
    IdIndicador = '1002000001'
    Idioma = 'es'
# Puede ser nacional [00], por entidad federativa [99] o por municipio [999], dependiendo de cada indicador.
    #Area_Geografica = '00'
# Puede consultarse solo el dato más reciente [true] o la serie histórica completa [false].
    Recientes = 'true'
# Corresponde a la fuente de diseminación de donde se obtendrán los datos consultados.
    Fuente_datos = 'BISE'
# Con él se identificará la edición [2.0] del servicio de provisión de datos.
    Version = '2.0'
# Se ofrece la información en 3 tipos de formatos: JSON [json], JSONP [jsonp] o XML [xml].
    Formato = 'json'

    url = f'https://inegi.org.mx/app/api/indicadores/desarrolladores/jsonxml/INDICATOR/{IdIndicador}/{Idioma}/{Area_Geografica}/{Recientes}/{Fuente_datos}/{Version}/{token}?type={Formato}'
    try:
        response = requests.get(url)
        #response.raise_for_status()
        data = response.json()

        # Extraemos el último dato disponible (Censo de Población)
        observaciones = data["Series"][0]["OBSERVATIONS"][0]

        poblacion = int(float(observaciones["OBS_VALUE"]))
        
        return poblacion

    except Exception as e:
        print(f"Error al consultar la región {Area_Geografica}: {e}")
        return None


# 1. Obtener los datos numéricos de la API
pob_total_mexico = consultar_poblacion("00")
pob_puebla = consultar_poblacion("21")
pob_veracruz = consultar_poblacion("30")

# 2. Imprimir los resultados numéricos obtenidos
print("DATOS OBTENIDOS DEL INEGI")
print(f"Población Total de México: {pob_total_mexico} habitantes")
print(f"Población de Puebla: {pob_puebla} habitantes")
print(f"Población de Veracruz: {pob_veracruz} habitantes\n")

# 3. Cálculo del ejercicio de probabilidad (Eventos mutuamente excluyentes)
if pob_total_mexico and pob_puebla and pob_veracruz:
    prob_puebla = pob_puebla / pob_total_mexico
    prob_veracruz = pob_veracruz / pob_total_mexico
    prob_union = prob_puebla + prob_veracruz

    print(f'Probabilidad de que alguien sea de Puebla si solo contemplamos la población de Puebla y Veracruz {prob_puebla/prob_union}')
```
