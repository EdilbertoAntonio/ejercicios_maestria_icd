<h1>Scaling and Data transformation</h1>
   <div style="text-align: justify">Data transformation refers to the process of converting raw data into a format that is suitable for analysis and modeling. After data transformation the data can be used to extract useful insights and knowledge.<br>
For some machine algorithms, it is important to have the same scale and we have to use scaling techniques for the data. One needs to rescale the data to ensure that it is on the same scale. There are many different approaches to doing this, for example, Min-Max scaling (sometimes referred to as normalization), and standardization.<br>
    <h2>min-max Scaling</h2>
Min-Max scaling is when your data is scaled linearly between a minimum and maximum value, often 0 and 1, with 0 corresponding with the lowest value in the column, and 1 with the largest. As it is a linear scaling while the values will change, the distribution will not. A Min-Max scaling is typically done via the following equation:<br>
$x_{sc}=\frac{x-x_{min}}{x_{max}-x_{min}}$<br>
In Python, one first needs to import MinMaxScaler from scikit learn's preprocessing module, scikit learn is the most commonly used machine learning library for python ($\color{red}{\text{Activity:}}\text{ Google scikitlearn module and explain its use}$). One then instantiates the MinMaxScaler() and fit it to your data. This tells the scaler how it should scale values when it performs the transformation. Finally, you need to actually transform the data with this new fitted scaler.
</div>

```python
#Example
from sklearn.preprocessing import MinMaxScaler
import pandas as pd

# create data
my_dat = pd.DataFrame([[10, 1], [3, 5], [0, 11], [12, 8]])

# scale features
scaler = MinMaxScaler()
model=scaler.fit(my_dat)
scaled_dat=model.transform(my_dat)

scaled_data1=scaler.fit_transform(my_dat)
# print scaled features
print(scaled_data1)
```

<h2>Standardization</h2>
<div style="text-align: justify">The other commonly used scaler is called standardization. Standardization finds the mean of the data and centers the distribution around it, calculating the number of standard deviations away from the mean each point is. These values (the number of standard deviations) are then used as your new values. This centers the data around 0 but technically has no limit to the maximum and minimum values.<br>
Standardization can be implemented in a similar way to the min-max scaler. One first imports StandardScaler from scikit-learn, instantiate and then fit it on your data. Once fitted you can apply it to your data.</div>

```python
from sklearn.preprocessing import StandardScaler
import pandas as pd
# create data
my_dat = pd.DataFrame([[10, 1], [3, 5], [0, 11], [12, 8]])

# compute required values
scaler = StandardScaler()
model = scaler.fit(my_dat)
scaled_dat = model.transform(my_dat)

# print scaled data
print(scaled_dat)
```

<h2>Log Transformation</h2>
<div style="text-align: justify">
Both normalization and min-max scaling are types of scalers, in other words the data remained in the same shape but was squashed or scaled. A log transformation on the other hand can be used to make highly skewed distributions less skewed.<br>
Although it effects your data quite differently, a log transformation is implemented in Python the same way you have implemented scalers. To use a log transform you first import PowerTransformer from sklearn's preprocessing module, then you fit it to your dataset, and once fitted you can transform your data. Log transformation is a type of power transformation, hence the name.<br>
    <h3>Note on Skewness and kurtosis</h3>
Skewness is a measure of the asymmetry of the probability distribution of a real-valued random variable about its mean. The skewness value can be positive, zero, negative.<br>
Kurtosis is a descriptive statistic used to help measure how data disperse between a distribution's center and tails, with larger values indicating a data distribution may have “heavy” tails that are thickly concentrated with observations or that are long with extreme observations.
</div>

```python
#Example
import numpy as np
import pandas as pd
from sklearn.preprocessing import PowerTransformer
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)
x = rng.standard_normal(1000)
y = pd.Series(np.exp(x))

p = PowerTransformer()
data_trans = p.fit_transform(y.to_frame())

plt.hist(data_trans, bins=30, edgecolor="black")
plt.xlabel
```
