```python
import numpy as np
import random

def quick_sort(lista):
    
    if len(lista) <= 1:
        return lista

    p = random.choice(lista)
    menores = [x for x in lista if x<p]
    iguales = [x for x in lista if x==p]
    mayores = [x for x in lista if x>p] 

    return quick_sort(menores) + iguales + quick_sort(mayores)

lista = [1,5,4,2,39,7,41,4,1,2,4,6,6,7]

quick_sort(lista)
```
