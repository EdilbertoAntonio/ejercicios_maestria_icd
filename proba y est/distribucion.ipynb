```python
import numpy as pd
import pandas as pd
import matplotlib.pyplot as plt
from bisect import bisect_right
```

```python
def dist_acumulada(valores, probabilidades):
    if len(probabilidades) != len(valores):
        print('error, longitudes distintas')
    
    if abs(sum(probabilidades)-1) > 0:
        print('error, las probabilidades no suman 1')

    if any(p<0 for p in probabilidades):
        print('error, las probabilidades deben ser positivas')

    pares=sorted(zip(valores, probabilidades))

    valores_orden=[x for x,_ in pares]
    acumuladas=[]
    suma=0

    for _,p in pares:
        suma+=p
        acumuladas.append(suma)

    def F(x):
        posicion = bisect_right(valores_orden,x)
        if posicion == 0:
            return 0
        return acumuladas[posicion-1]
    return F
```

```python
valores = [3,0,2]
probabilidades = [0.5, 0.3, 0.2]

F = dist_acumulada(valores, probabilidades)
print(F(3))
```

```python
plt.step(sorted(valores), [F(x) for x in sorted(valores)], where='post', color='green')
```

```python
plt.stem(sorted(valores), [F(x) for x in sorted(valores)])
```
